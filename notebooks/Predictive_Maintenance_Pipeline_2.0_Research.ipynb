```python
# ============================================================
# CORRECTED PIPELINE 2.0
# STAGE 1 — RAW DATA AUDIT AND EXPLORATORY DATA ANALYSIS
# ============================================================

import os
import json
import warnings
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

warnings.filterwarnings("ignore")

print("=" * 80)
print("CORRECTED PIPELINE 2.0")
print("STAGE 1 — RAW DATA AUDIT AND EXPLORATORY DATA ANALYSIS")
print("=" * 80)

# ------------------------------------------------------------
# 1. PATHS
# ------------------------------------------------------------

raw_data_path = r"D:\Data_Workspace\Capstone_Project\NLGN_PLC_Predictive_Analystic\data\01_raw\NLNG_Dirty_Data.csv"

corrected_root = r"D:\Data_Workspace\Capstone_Project\NLGN_PLC_Predictive_Analystic\data\04_corrected_pipeline"

stage1_dir = os.path.join(corrected_root, "stage_1_audit_eda")

os.makedirs(stage1_dir, exist_ok=True)

print(f"\nRaw dataset:")
print(raw_data_path)

print(f"\nCorrected pipeline output:")
print(corrected_root)


# ------------------------------------------------------------
# 2. LOAD RAW DATA
# ------------------------------------------------------------

raw_df = pd.read_csv(raw_data_path)

# Standardise headers only.
# IMPORTANT:
# This does NOT modify values.
raw_df.columns = (
    raw_df.columns
    .str.strip()
    .str.lower()
)

print("\n" + "=" * 80)
print("RAW DATASET")
print("=" * 80)

print(f"Rows:    {len(raw_df):,}")
print(f"Columns: {len(raw_df.columns):,}")
print(f"Memory:  {raw_df.memory_usage(deep=True).sum() / 1024**2:.2f} MB")

print("\nFirst 5 rows:")
display(raw_df.head())


# ------------------------------------------------------------
# 3. SCHEMA AUDIT
# ------------------------------------------------------------

schema_report = pd.DataFrame({
    "column": raw_df.columns,
    "dtype": [str(raw_df[c].dtype) for c in raw_df.columns],
    "missing_count": [raw_df[c].isna().sum() for c in raw_df.columns],
    "missing_pct": [
        raw_df[c].isna().mean() * 100
        for c in raw_df.columns
    ],
    "unique_values": [
        raw_df[c].nunique(dropna=False)
        for c in raw_df.columns
    ]
})

schema_report = schema_report.sort_values(
    "missing_pct",
    ascending=False
)

schema_report.to_csv(
    os.path.join(stage1_dir, "01_schema_audit.csv"),
    index=False
)

print("\nSchema audit saved.")


# ------------------------------------------------------------
# 4. DUPLICATE AUDIT
# ------------------------------------------------------------

exact_duplicates = raw_df.duplicated().sum()

if {"equipment_id", "timestamp"}.issubset(raw_df.columns):

    equipment_timestamp_duplicates = raw_df.duplicated(
        subset=["equipment_id", "timestamp"],
        keep=False
    ).sum()

else:
    equipment_timestamp_duplicates = np.nan

duplicate_report = pd.DataFrame({
    "audit_item": [
        "exact_duplicate_rows",
        "equipment_timestamp_duplicate_rows"
    ],
    "count": [
        exact_duplicates,
        equipment_timestamp_duplicates
    ]
})

duplicate_report.to_csv(
    os.path.join(stage1_dir, "02_duplicate_audit.csv"),
    index=False
)

print("\nDuplicate audit:")
display(duplicate_report)


# ------------------------------------------------------------
# 5. TIMESTAMP AUDIT
# ------------------------------------------------------------

raw_df["timestamp"] = pd.to_datetime(
    raw_df["timestamp"],
    errors="coerce"
)

timestamp_report = pd.DataFrame({
    "metric": [
        "invalid_or_missing_timestamps",
        "minimum_timestamp",
        "maximum_timestamp",
        "unique_timestamps"
    ],
    "value": [
        raw_df["timestamp"].isna().sum(),
        raw_df["timestamp"].min(),
        raw_df["timestamp"].max(),
        raw_df["timestamp"].nunique()
    ]
})

timestamp_report.to_csv(
    os.path.join(stage1_dir, "03_timestamp_audit.csv"),
    index=False
)

print("\nTimestamp audit:")
display(timestamp_report)


# ------------------------------------------------------------
# 6. EQUIPMENT AND TRAIN COVERAGE
# ------------------------------------------------------------

if "equipment_id" in raw_df.columns:

    equipment_coverage = (
        raw_df["equipment_id"]
        .value_counts()
        .rename_axis("equipment_id")
        .reset_index(name="record_count")
    )

    equipment_coverage.to_csv(
        os.path.join(stage1_dir, "04_equipment_coverage.csv"),
        index=False
    )

if "train" in raw_df.columns:

    train_coverage = (
        raw_df["train"]
        .value_counts()
        .rename_axis("train")
        .reset_index(name="record_count")
    )

    train_coverage.to_csv(
        os.path.join(stage1_dir, "05_train_coverage.csv"),
        index=False
    )

print("\nEquipment assets:",
      raw_df["equipment_id"].nunique()
      if "equipment_id" in raw_df.columns else "N/A")

print("Trains:",
      raw_df["train"].nunique()
      if "train" in raw_df.columns else "N/A")


# ------------------------------------------------------------
# 7. MISSING VALUE AUDIT
# ------------------------------------------------------------

missing_report = pd.DataFrame({
    "column": raw_df.columns,
    "missing_count": raw_df.isna().sum().values,
    "missing_percentage": (
        raw_df.isna().mean().values * 100
    )
})

missing_report = missing_report.sort_values(
    "missing_percentage",
    ascending=False
)

missing_report.to_csv(
    os.path.join(stage1_dir, "06_missing_value_audit.csv"),
    index=False
)

print("\n" + "=" * 80)
print("MISSING VALUE AUDIT")
print("=" * 80)

display(
    missing_report[
        missing_report["missing_count"] > 0
    ]
)


# ------------------------------------------------------------
# 8. TARGET AUDIT — FAILURE WITHIN 24 HOURS
# ------------------------------------------------------------

classification_target = "failure_within_24h"

if classification_target in raw_df.columns:

    failure_counts = (
        raw_df[classification_target]
        .value_counts(dropna=False)
        .sort_index()
    )

    failure_distribution = pd.DataFrame({
        "count": failure_counts,
        "percentage": (
            failure_counts / len(raw_df) * 100
        )
    })

    failure_distribution.to_csv(
        os.path.join(
            stage1_dir,
            "07_failure_24h_distribution.csv"
        )
    )

    print("\n" + "=" * 80)
    print("FAILURE WITHIN 24 HOURS")
    print("=" * 80)

    display(failure_distribution)


# ------------------------------------------------------------
# 9. RUL AUDIT
# ------------------------------------------------------------

rul_target = "rul_days"

if rul_target in raw_df.columns:

    rul_summary = raw_df[rul_target].describe()

    rul_audit = pd.DataFrame({
        "metric": [
            "missing",
            "negative",
            "mean",
            "median",
            "minimum",
            "maximum"
        ],
        "value": [
            raw_df[rul_target].isna().sum(),
            (raw_df[rul_target] < 0).sum(),
            raw_df[rul_target].mean(),
            raw_df[rul_target].median(),
            raw_df[rul_target].min(),
            raw_df[rul_target].max()
        ]
    })

    rul_audit.to_csv(
        os.path.join(stage1_dir, "08_rul_audit.csv"),
        index=False
    )

    print("\n" + "=" * 80)
    print("RUL AUDIT")
    print("=" * 80)

    display(rul_audit)


# ------------------------------------------------------------
# 10. PHYSICAL VALIDITY AUDIT
# ------------------------------------------------------------

physical_rules = {
    "load_factor": (0, 1),
    "wear_level": (0, 1),
    "lubrication_health_index": (0, 100),
    "oil_pressure": (0, np.inf),
    "oil_particles_ppm": (0, np.inf),
    "overall_vibration": (0, np.inf),
    "vibration": (0, np.inf),
    "rpm": (0, np.inf),
    "hours_since_maint": (0, np.inf),
    "cumulative_op_hours": (0, np.inf),
    "rul_days": (0, np.inf)
}

physical_results = []

for col, (lower, upper) in physical_rules.items():

    if col not in raw_df.columns:
        continue

    invalid_mask = (
        (raw_df[col] < lower) |
        (raw_df[col] > upper)
    ) & raw_df[col].notna()

    physical_results.append({
        "variable": col,
        "lower_bound": lower,
        "upper_bound": upper,
        "invalid_count": int(invalid_mask.sum()),
        "invalid_pct": float(
            invalid_mask.mean() * 100
        )
    })

physical_report = pd.DataFrame(physical_results)

physical_report.to_csv(
    os.path.join(stage1_dir, "09_physical_validity_audit.csv"),
    index=False
)

print("\n" + "=" * 80)
print("PHYSICAL VALIDITY AUDIT")
print("=" * 80)

display(physical_report)


# ------------------------------------------------------------
# 11. OUTLIER AUDIT — IQR
# ------------------------------------------------------------

numeric_columns = raw_df.select_dtypes(
    include=np.number
).columns

outlier_results = []

for col in numeric_columns:

    series = raw_df[col].dropna()

    if len(series) == 0:
        continue

    q1 = series.quantile(0.25)
    q3 = series.quantile(0.75)

    iqr = q3 - q1

    lower = q1 - 1.5 * iqr
    upper = q3 + 1.5 * iqr

    mask = (
        (raw_df[col] < lower) |
        (raw_df[col] > upper)
    )

    outlier_results.append({
        "variable": col,
        "lower_iqr": lower,
        "upper_iqr": upper,
        "outlier_count": int(mask.sum()),
        "outlier_pct": float(mask.mean() * 100)
    })

outlier_report = (
    pd.DataFrame(outlier_results)
    .sort_values(
        "outlier_pct",
        ascending=False
    )
)

outlier_report.to_csv(
    os.path.join(stage1_dir, "10_outlier_audit_iqr.csv"),
    index=False
)

print("\n" + "=" * 80)
print("TOP OUTLIER VARIABLES")
print("=" * 80)

display(outlier_report.head(15))


# ------------------------------------------------------------
# 12. LEAKAGE SCREENING
# ------------------------------------------------------------

potential_leakage = [
    "failure_within_72h",
    "failure_within_7d",
    "rul_days",
    "rul_censored",
    "failure_rate"
]

leakage_report = pd.DataFrame({
    "variable": potential_leakage,
    "present": [
        c in raw_df.columns
        for c in potential_leakage
    ],
    "treatment": [
        "EXCLUDE — future failure target"
        if c in ["failure_within_72h", "failure_within_7d"]
        else
        "EXCLUDE — regression target"
        if c in ["rul_days", "rul_censored"]
        else
        "EXCLUDE CONSERVATIVELY — leakage provenance requires verification"
        if c == "failure_rate"
        else "Not present"
        for c in potential_leakage
    ]
})

leakage_report.to_csv(
    os.path.join(stage1_dir, "11_leakage_screening.csv"),
    index=False
)

print("\n" + "=" * 80)
print("LEAKAGE SCREENING")
print("=" * 80)

display(leakage_report)


# ------------------------------------------------------------
# 13. DESCRIPTIVE STATISTICS
# ------------------------------------------------------------

numeric_profile = (
    raw_df
    .select_dtypes(include=np.number)
    .describe()
    .T
)

numeric_profile["skewness"] = (
    raw_df
    .select_dtypes(include=np.number)
    .skew()
)

numeric_profile.to_csv(
    os.path.join(
        stage1_dir,
        "12_raw_statistical_profile.csv"
    )
)

print("\n" + "=" * 80)
print("RAW NUMERICAL PROFILE")
print("=" * 80)

display(numeric_profile)


# ------------------------------------------------------------
# 14. RUL DISTRIBUTION
# ------------------------------------------------------------

if rul_target in raw_df.columns:

    plt.figure(figsize=(8, 5))

    raw_df[rul_target].dropna().plot(
        kind="hist",
        bins=50
    )

    plt.xlabel("RUL (Days)")
    plt.ylabel("Frequency")
    plt.title("Raw RUL Distribution")
    plt.tight_layout()

    plt.savefig(
        os.path.join(
            stage1_dir,
            "eda_rul_distribution.png"
        ),
        dpi=300
    )

    plt.show()
    plt.close()


# ------------------------------------------------------------
# 15. FAILURE DISTRIBUTION
# ------------------------------------------------------------

if classification_target in raw_df.columns:

    failure_counts = (
        raw_df[classification_target]
        .value_counts()
        .sort_index()
    )

    plt.figure(figsize=(7, 5))

    failure_counts.plot(
        kind="bar"
    )

    plt.xlabel("Failure Within 24 Hours")
    plt.ylabel("Number of Observations")
    plt.title("24-Hour Failure Class Distribution")
    plt.xticks(rotation=0)

    plt.tight_layout()

    plt.savefig(
        os.path.join(
            stage1_dir,
            "eda_failure_distribution.png"
        ),
        dpi=300
    )

    plt.show()
    plt.close()


# ------------------------------------------------------------
# 16. KEY SENSOR DISTRIBUTIONS
# ------------------------------------------------------------

key_sensors = [
    "overall_vibration",
    "vibration",
    "oil_particles_ppm",
    "oil_pressure",
    "lubrication_health_index",
    "load_factor",
    "wear_level"
]

available_sensors = [
    c for c in key_sensors
    if c in raw_df.columns
]

for col in available_sensors:

    plt.figure(figsize=(7, 5))

    raw_df[col].dropna().plot(
        kind="hist",
        bins=50
    )

    plt.xlabel(col)
    plt.ylabel("Frequency")
    plt.title(f"Distribution of {col}")

    plt.tight_layout()

    plt.savefig(
        os.path.join(
            stage1_dir,
            f"eda_{col}_distribution.png"
        ),
        dpi=300
    )

    plt.show()
    plt.close()


# ------------------------------------------------------------
# 17. SENSOR VALUES BY FAILURE STATUS
# ------------------------------------------------------------

if classification_target in raw_df.columns:

    sensor_failure_summary = []

    for col in available_sensors:

        normal = raw_df.loc[
            raw_df[classification_target] == 0,
            col
        ].dropna()

        failure = raw_df.loc[
            raw_df[classification_target] == 1,
            col
        ].dropna()

        sensor_failure_summary.append({
            "sensor": col,
            "normal_n": len(normal),
            "failure_n": len(failure),
            "normal_median": normal.median(),
            "failure_median": failure.median(),
            "median_difference": (
                failure.median() - normal.median()
            )
        })

    sensor_failure_report = pd.DataFrame(
        sensor_failure_summary
    )

    sensor_failure_report.to_csv(
        os.path.join(
            stage1_dir,
            "13_sensor_failure_comparison.csv"
        ),
        index=False
    )

    print("\n" + "=" * 80)
    print("SENSOR VALUES BY FAILURE STATUS")
    print("=" * 80)

    display(sensor_failure_report)


# ------------------------------------------------------------
# 18. STAGE 1 SUMMARY
# ------------------------------------------------------------

stage1_summary = {
    "rows": int(len(raw_df)),
    "columns": int(len(raw_df.columns)),
    "exact_duplicates": int(exact_duplicates),
    "equipment_timestamp_duplicates": (
        int(equipment_timestamp_duplicates)
        if not pd.isna(equipment_timestamp_duplicates)
        else None
    ),
    "missing_value_columns": int(
        (raw_df.isna().sum() > 0).sum()
    ),
    "equipment_assets": (
        int(raw_df["equipment_id"].nunique())
        if "equipment_id" in raw_df.columns
        else None
    ),
    "trains": (
        int(raw_df["train"].nunique())
        if "train" in raw_df.columns
        else None
    )
}

with open(
    os.path.join(stage1_dir, "stage1_summary.json"),
    "w"
) as f:

    json.dump(
        stage1_summary,
        f,
        indent=4
    )

print("\n" + "=" * 80)
print("STAGE 1 COMPLETE")
print("=" * 80)

print(json.dumps(stage1_summary, indent=4))
```

```python
# ============================================================
# CORRECTED PIPELINE 2.0
# STAGE 2 — DATA CLEANING, LEAKAGE CONTROL AND TEMPORAL SPLIT
# ============================================================

import os
import json
import numpy as np
import pandas as pd

from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler

print("=" * 80)
print("STAGE 2 — CORRECTED DATA CLEANING AND TEMPORAL PARTITION")
print("=" * 80)


# ------------------------------------------------------------
# 1. PATHS
# ------------------------------------------------------------

raw_data_path = r"D:\Data_Workspace\Capstone_Project\NLGN_PLC_Predictive_Analystic\data\01_raw\NLNG_Dirty_Data.csv"

corrected_root = r"D:\Data_Workspace\Capstone_Project\NLGN_PLC_Predictive_Analystic\data\04_corrected_pipeline"

stage2_dir = os.path.join(
    corrected_root,
    "stage_2_cleaning"
)

os.makedirs(stage2_dir, exist_ok=True)


# ------------------------------------------------------------
# 2. LOAD RAW DATA AGAIN
# ------------------------------------------------------------

df = pd.read_csv(raw_data_path)

df.columns = (
    df.columns
    .str.strip()
    .str.lower()
)

df["timestamp"] = pd.to_datetime(
    df["timestamp"],
    errors="coerce"
)

print(f"Original rows: {len(df):,}")


# ------------------------------------------------------------
# 3. REMOVE INVALID TIMESTAMPS
# ------------------------------------------------------------

invalid_timestamp_count = df["timestamp"].isna().sum()

df = df.dropna(
    subset=["timestamp"]
).copy()


# ------------------------------------------------------------
# 4. REMOVE EXACT DUPLICATES
# ------------------------------------------------------------

exact_duplicates_before = df.duplicated().sum()

df = df.drop_duplicates(
    keep="first"
).copy()

print(
    f"Exact duplicate rows removed: "
    f"{exact_duplicates_before:,}"
)


# ------------------------------------------------------------
# 5. SORT BY EQUIPMENT + TIMESTAMP
# ------------------------------------------------------------

df = df.sort_values(
    ["equipment_id", "timestamp", "row_id"],
    kind="mergesort"
).reset_index(drop=True)


# ------------------------------------------------------------
# 6. EQUIPMENT + TIMESTAMP DUPLICATES
# ------------------------------------------------------------

equipment_timestamp_duplicate_rows = df.duplicated(
    subset=["equipment_id", "timestamp"],
    keep=False
).sum()

print(
    "Equipment + timestamp duplicate rows remaining: "
    f"{equipment_timestamp_duplicate_rows:,}"
)

# We do NOT silently delete these.
# We create a deterministic resolution report.

duplicate_group_sizes = (
    df.groupby(
        ["equipment_id", "timestamp"],
        dropna=False
    )
    .size()
    .reset_index(name="records_at_same_timestamp")
)

duplicate_groups = duplicate_group_sizes[
    duplicate_group_sizes["records_at_same_timestamp"] > 1
]

duplicate_groups.to_csv(
    os.path.join(
        stage2_dir,
        "equipment_timestamp_duplicate_groups.csv"
    ),
    index=False
)

# For modelling, one row per equipment/timestamp is required.
# We retain the first row after deterministic Row_ID ordering.
#
# This treatment is explicitly recorded rather than hidden.

df = df.drop_duplicates(
    subset=["equipment_id", "timestamp"],
    keep="first"
).reset_index(drop=True)

print(
    f"Rows after equipment/timestamp deduplication: "
    f"{len(df):,}"
)


# ------------------------------------------------------------
# 7. PHYSICAL VALIDITY TREATMENT
# ------------------------------------------------------------

# IMPORTANT:
# These rules are based on the audit and are deliberately explicit.
# Outliers are NOT removed simply because they are statistically unusual.

physical_bounds = {
    "load_factor": (0, 1),
    "wear_level": (0, 1),
    "lubrication_health_index": (0, 100),
    "oil_pressure": (0, np.inf),
    "oil_particles_ppm": (0, np.inf),
    "overall_vibration": (0, np.inf),
    "vibration": (0, np.inf),
    "rpm": (0, np.inf),
    "hours_since_maint": (0, np.inf),
    "cumulative_op_hours": (0, np.inf),
    "rul_days": (0, np.inf)
}

physical_cleaning_log = []

for col, (lower, upper) in physical_bounds.items():

    if col not in df.columns:
        continue

    invalid_mask = (
        (df[col] < lower) |
        (df[col] > upper)
    ) & df[col].notna()

    invalid_count = int(invalid_mask.sum())

    if invalid_count > 0:
        df.loc[invalid_mask, col] = np.nan

    physical_cleaning_log.append({
        "variable": col,
        "lower_bound": lower,
        "upper_bound": upper,
        "converted_to_nan": invalid_count
    })

physical_cleaning_report = pd.DataFrame(
    physical_cleaning_log
)

physical_cleaning_report.to_csv(
    os.path.join(
        stage2_dir,
        "physical_cleaning_log.csv"
    ),
    index=False
)

display(physical_cleaning_report)


# ------------------------------------------------------------
# 8. DO NOT IMPUTE TARGET VARIABLES
# ------------------------------------------------------------

target_columns = [
    "failure_within_24h",
    "failure_within_72h",
    "failure_within_7d",
    "rul_days",
    "rul_censored"
]

print("\nTarget columns protected from imputation:")

for col in target_columns:
    if col in df.columns:
        print(f"  - {col}")


# ------------------------------------------------------------
# 9. CONSERVATIVE LEAKAGE CONTROL
# ------------------------------------------------------------

# These variables are excluded from the modelling feature contract.

excluded_features = [
    "row_id",
    "timestamp",
    "train",
    "equipment_id",
    "equipment_name",
    "equipment_type",
    "area",
    "manufacturer",
    "criticality",

    # RUL target / censoring information
    "rul_days",
    "rul_censored",

    # Failure targets
    "failure_within_24h",
    "failure_within_72h",
    "failure_within_7d",

    # Conservative leakage exclusion
    # until its provenance is verified.
    "failure_rate"
]

excluded_features = [
    c for c in excluded_features
    if c in df.columns
]

print("\nExcluded modelling variables:")
for c in excluded_features:
    print(f"  - {c}")


# ------------------------------------------------------------
# 10. DEFINE NUMERICAL FEATURE CONTRACT
# ------------------------------------------------------------

feature_df = df.drop(
    columns=excluded_features,
    errors="ignore"
)

feature_df = feature_df.select_dtypes(
    include=[np.number]
)

feature_columns = feature_df.columns.tolist()

print(
    f"\nFinal numerical feature count: "
    f"{len(feature_columns)}"
)

print("\nFeatures:")
for i, col in enumerate(feature_columns, 1):
    print(f"{i:02d}. {col}")


# ------------------------------------------------------------
# 11. SAVE FEATURE CONTRACT
# ------------------------------------------------------------

feature_contract = {
    "classification_target": "failure_within_24h",
    "regression_target": "rul_days",
    "excluded_features": excluded_features,
    "numerical_features": feature_columns,
    "failure_rate_status": "EXCLUDED_PENDING_PROVENANCE_REVIEW"
}

with open(
    os.path.join(
        stage2_dir,
        "feature_contract.json"
    ),
    "w"
) as f:

    json.dump(
        feature_contract,
        f,
        indent=4
    )


# ------------------------------------------------------------
# 12. GLOBAL CHRONOLOGICAL BOUNDARIES
# ------------------------------------------------------------

min_time = df["timestamp"].min()
max_time = df["timestamp"].max()

print("\nTime range:")
print("Start:", min_time)
print("End:  ", max_time)


# ------------------------------------------------------------
# 13. 64 / 16 / 20 TEMPORAL SPLIT
# ------------------------------------------------------------

# We use timestamp boundaries rather than arbitrary row positions.
#
# 64% DEVELOPMENT
# 16% VALIDATION
# 20% FINAL HOLDOUT TEST

unique_times = np.sort(
    df["timestamp"].dropna().unique()
)

n_times = len(unique_times)

train_time_idx = int(
    n_times * 0.64
)

validation_time_idx = int(
    n_times * 0.80
)

train_end = unique_times[
    train_time_idx - 1
]

validation_end = unique_times[
    validation_time_idx - 1
]

print("\nTemporal boundaries:")
print("Training ends:   ", train_end)
print("Validation ends: ", validation_end)
print("Holdout begins:  ", validation_end)


# ------------------------------------------------------------
# 14. CREATE SPLIT LABEL
# ------------------------------------------------------------

df["data_partition"] = np.select(
    [
        df["timestamp"] <= train_end,
        df["timestamp"] <= validation_end
    ],
    [
        "train",
        "validation"
    ],
    default="test"
)


# ------------------------------------------------------------
# 15. VERIFY TEMPORAL ORDER
# ------------------------------------------------------------

train_df = df[
    df["data_partition"] == "train"
].copy()

validation_df = df[
    df["data_partition"] == "validation"
].copy()

test_df = df[
    df["data_partition"] == "test"
].copy()

print("\nPartition sizes:")

partition_summary = pd.DataFrame({
    "partition": [
        "train",
        "validation",
        "test"
    ],
    "rows": [
        len(train_df),
        len(validation_df),
        len(test_df)
    ],
    "percentage": [
        len(train_df) / len(df) * 100,
        len(validation_df) / len(df) * 100,
        len(test_df) / len(df) * 100
    ],
    "start_time": [
        train_df["timestamp"].min(),
        validation_df["timestamp"].min(),
        test_df["timestamp"].min()
    ],
    "end_time": [
        train_df["timestamp"].max(),
        validation_df["timestamp"].max(),
        test_df["timestamp"].max()
    ]
})

display(partition_summary)


# ------------------------------------------------------------
# 16. TEMPORAL LEAKAGE CHECK
# ------------------------------------------------------------

assert (
    train_df["timestamp"].max()
    <
    validation_df["timestamp"].min()
)

assert (
    validation_df["timestamp"].max()
    <
    test_df["timestamp"].min()
)

print("\nTemporal boundary audit: PASSED")


# ------------------------------------------------------------
# 17. SAVE CLEANED DATA
# ------------------------------------------------------------

cleaned_path = os.path.join(
    stage2_dir,
    "NLNG_cleaned_leakage_controlled.csv"
)

df.to_csv(
    cleaned_path,
    index=False
)

print("\nCleaned dataset saved:")
print(cleaned_path)


# ------------------------------------------------------------
# 18. SAVE PARTITIONS
# ------------------------------------------------------------

train_path = os.path.join(
    stage2_dir,
    "train_raw_partition.csv"
)

validation_path = os.path.join(
    stage2_dir,
    "validation_raw_partition.csv"
)

test_path = os.path.join(
    stage2_dir,
    "test_raw_partition.csv"
)

train_df.to_csv(train_path, index=False)
validation_df.to_csv(validation_path, index=False)
test_df.to_csv(test_path, index=False)


# ------------------------------------------------------------
# 19. SAVE SPLIT METADATA
# ------------------------------------------------------------

split_metadata = {
    "strategy": "chronological_timestamp_split",
    "train_percentage_target": 64,
    "validation_percentage_target": 16,
    "test_percentage_target": 20,
    "train_end": str(train_end),
    "validation_end": str(validation_end),
    "test_start": str(test_df["timestamp"].min()),
    "train_rows": int(len(train_df)),
    "validation_rows": int(len(validation_df)),
    "test_rows": int(len(test_df))
}

with open(
    os.path.join(
        stage2_dir,
        "temporal_split_metadata.json"
    ),
    "w"
) as f:

    json.dump(
        split_metadata,
        f,
        indent=4
    )


print("\n" + "=" * 80)
print("STAGE 2 COMPLETE")
print("=" * 80)
```

```python
# ============================================================
# CORRECTED PIPELINE 2.0
# STAGE 3 — MACHINE LEARNING
# ============================================================

import os
import json
import joblib
import numpy as np
import pandas as pd

from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler

from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import (
    RandomForestClassifier,
    GradientBoostingClassifier,
    RandomForestRegressor,
    GradientBoostingRegressor
)

from sklearn.metrics import (
    average_precision_score,
    roc_auc_score,
    mean_squared_error,
    r2_score
)

print("=" * 80)
print("STAGE 3 — CORRECTED MACHINE LEARNING PIPELINE")
print("=" * 80)


# ------------------------------------------------------------
# 1. PATHS
# ------------------------------------------------------------

corrected_root = r"D:\Data_Workspace\Capstone_Project\NLGN_PLC_Predictive_Analystic\data\04_corrected_pipeline"

stage2_dir = os.path.join(
    corrected_root,
    "stage_2_cleaning"
)

stage3_dir = os.path.join(
    corrected_root,
    "stage_3_ml"
)

os.makedirs(stage3_dir, exist_ok=True)


# ------------------------------------------------------------
# 2. LOAD DATA
# ------------------------------------------------------------

train_df = pd.read_csv(
    os.path.join(
        stage2_dir,
        "train_raw_partition.csv"
    )
)

validation_df = pd.read_csv(
    os.path.join(
        stage2_dir,
        "validation_raw_partition.csv"
    )
)

test_df = pd.read_csv(
    os.path.join(
        stage2_dir,
        "test_raw_partition.csv"
    )
)

# Load feature contract
with open(
    os.path.join(
        stage2_dir,
        "feature_contract.json"
    ),
    "r"
) as f:

    feature_contract = json.load(f)

feature_columns = feature_contract[
    "numerical_features"
]

classification_target = (
    feature_contract["classification_target"]
)

regression_target = (
    feature_contract["regression_target"]
)

print(
    f"Features: {len(feature_columns)}"
)


# ------------------------------------------------------------
# 3. CLASSIFICATION DATA
# ------------------------------------------------------------

X_train_c = train_df[
    feature_columns
].copy()

X_val_c = validation_df[
    feature_columns
].copy()

X_test_c = test_df[
    feature_columns
].copy()

y_train_c = train_df[
    classification_target
].astype(int)

y_val_c = validation_df[
    classification_target
].astype(int)

y_test_c = test_df[
    classification_target
].astype(int
)

print("\nClassification class distribution:")

print(
    "Train:",
    y_train_c.value_counts(normalize=True)
)

print(
    "Validation:",
    y_val_c.value_counts(normalize=True)
)

print(
    "Test:",
    y_test_c.value_counts(normalize=True)
)


# ------------------------------------------------------------
# 4. CLASSIFICATION MODELS
# ------------------------------------------------------------

classification_models = {

    "Logistic Regression": Pipeline([
        (
            "imputer",
            SimpleImputer(
                strategy="median",
                add_indicator=True
            )
        ),
        (
            "scaler",
            StandardScaler()
        ),
        (
            "model",
            LogisticRegression(
                max_iter=1500,
                class_weight="balanced",
                random_state=42
            )
        )
    ]),

    "Random Forest Classifier": Pipeline([
        (
            "imputer",
            SimpleImputer(
                strategy="median",
                add_indicator=True
            )
        ),
        (
            "model",
            RandomForestClassifier(
                n_estimators=100,
                class_weight="balanced",
                random_state=42,
                n_jobs=-1
            )
        )
    ]),

    "Gradient Boosting Classifier": Pipeline([
        (
            "imputer",
            SimpleImputer(
                strategy="median",
                add_indicator=True
            )
        ),
        (
            "model",
            GradientBoostingClassifier(
                n_estimators=100,
                random_state=42
            )
        )
    ])
}


# ------------------------------------------------------------
# 5. TRAIN CLASSIFIERS
# ------------------------------------------------------------

classification_results = []

for name, model in classification_models.items():

    print("\n" + "-" * 70)
    print(f"Training: {name}")
    print("-" * 70)

    model.fit(
        X_train_c,
        y_train_c
    )

    # Validation
    val_probs = model.predict_proba(
        X_val_c
    )[:, 1]

    val_pr_auc = average_precision_score(
        y_val_c,
        val_probs
    )

    val_roc_auc = roc_auc_score(
        y_val_c,
        val_probs
    )

    # Test metrics are calculated but NOT used
    # for champion selection.
    test_probs = model.predict_proba(
        X_test_c
    )[:, 1]

    test_pr_auc = average_precision_score(
        y_test_c,
        test_probs
    )

    test_roc_auc = roc_auc_score(
        y_test_c,
        test_probs
    )

    classification_results.append({
        "Model": name,
        "Task": "Classification",
        "Validation_PR_AUC": val_pr_auc,
        "Validation_ROC_AUC": val_roc_auc,
        "Holdout_PR_AUC": test_pr_auc,
        "Holdout_ROC_AUC": test_roc_auc
    })

    filename = (
        name.lower()
        .replace(" ", "_")
        .replace("-", "")
    )

    joblib.dump(
        model,
        os.path.join(
            stage3_dir,
            f"{filename}.pkl"
        )
    )

classification_results_df = pd.DataFrame(
    classification_results
)

classification_results_df = (
    classification_results_df
    .sort_values(
        "Validation_PR_AUC",
        ascending=False
    )
)

print("\n" + "=" * 80)
print("CLASSIFICATION RESULTS")
print("=" * 80)

display(classification_results_df)

classification_results_df.to_csv(
    os.path.join(
        stage3_dir,
        "classification_results.csv"
    ),
    index=False
)


# ------------------------------------------------------------
# 6. RUL REGRESSION DATA
# ------------------------------------------------------------

train_rul = train_df[
    train_df[regression_target].notna()
].copy()

validation_rul = validation_df[
    validation_df[regression_target].notna()
].copy()

test_rul = test_df[
    test_df[regression_target].notna()
].copy()

X_train_r = train_rul[
    feature_columns
].copy()

X_val_r = validation_rul[
    feature_columns
].copy()

X_test_r = test_rul[
    feature_columns
].copy()

y_train_r = train_rul[
    regression_target
].astype(float)

y_val_r = validation_rul[
    regression_target
].astype(float)

y_test_r = test_rul[
    regression_target
].astype(float)

print("\nRUL observations:")
print("Train:", len(y_train_r))
print("Validation:", len(y_val_r))
print("Test:", len(y_test_r))


# ------------------------------------------------------------
# 7. REGRESSION MODELS
# ------------------------------------------------------------

regression_models = {

    "Random Forest Regressor": Pipeline([
        (
            "imputer",
            SimpleImputer(
                strategy="median",
                add_indicator=True
            )
        ),
        (
            "model",
            RandomForestRegressor(
                n_estimators=100,
                max_depth=15,
                random_state=42,
                n_jobs=-1
            )
        )
    ]),

    "Gradient Boosting Regressor": Pipeline([
        (
            "imputer",
            SimpleImputer(
                strategy="median",
                add_indicator=True
            )
        ),
        (
            "model",
            GradientBoostingRegressor(
                n_estimators=100,
                max_depth=5,
                random_state=42
            )
        )
    ])
}


# ------------------------------------------------------------
# 8. TRAIN REGRESSORS
# ------------------------------------------------------------

regression_results = []

for name, model in regression_models.items():

    print("\n" + "-" * 70)
    print(f"Training: {name}")
    print("-" * 70)

    model.fit(
        X_train_r,
        y_train_r
    )

    val_preds = model.predict(
        X_val_r
    )

    test_preds = model.predict(
        X_test_r
    )

    val_rmse = np.sqrt(
        mean_squared_error(
            y_val_r,
            val_preds
        )
    )

    val_r2 = r2_score(
        y_val_r,
        val_preds
    )

    test_rmse = np.sqrt(
        mean_squared_error(
            y_test_r,
            test_preds
        )
    )

    test_r2 = r2_score(
        y_test_r,
        test_preds
    )

    regression_results.append({
        "Model": name,
        "Task": "Regression",
        "Validation_RMSE_Days": val_rmse,
        "Validation_R2": val_r2,
        "Holdout_RMSE_Days": test_rmse,
        "Holdout_R2": test_r2
    })

    filename = (
        name.lower()
        .replace(" ", "_")
        .replace("-", "")
    )

    joblib.dump(
        model,
        os.path.join(
            stage3_dir,
            f"{filename}.pkl"
        )
    )

regression_results_df = pd.DataFrame(
    regression_results
)

regression_results_df = (
    regression_results_df
    .sort_values(
        "Validation_RMSE_Days",
        ascending=True
    )
)

print("\n" + "=" * 80)
print("REGRESSION RESULTS")
print("=" * 80)

display(regression_results_df)

regression_results_df.to_csv(
    os.path.join(
        stage3_dir,
        "regression_results.csv"
    ),
    index=False
)


print("\n" + "=" * 80)
print("STAGE 3 COMPLETE")
print("=" * 80)
```

```python
# =============================================================================
# STAGE 4 — CORRECTED DEEP LEARNING PIPELINE
# =============================================================================
#
# Purpose:
#   1. Deep Learning Classification using MLP
#   2. Deep Learning RUL Regression using LSTM
#
# Important correction:
#   The previous LSTM implementation flattened all sequence windows before
#   applying SimpleImputer(strategy="median"). This created a very large
#   intermediate matrix and caused a MemoryError.
#
#   This version:
#       - imputes/scales the feature matrix BEFORE sequence construction
#       - uses float32 instead of float64
#       - constructs LSTM sequences lazily in batches
#       - preserves chronological train/validation/test boundaries
#
# =============================================================================

# =============================================================================
# STAGE 4 OUTPUT DIRECTORY
# =============================================================================

output_dir = r"D:\Data_Workspace\Capstone_Project\NLGN_PLC_Predictive_Analystic\data\04_corrected_pipeline\stage_4_deep_learning"

os.makedirs(output_dir, exist_ok=True)

print(f"Stage 4 output directory:")
print(output_dir)




print("\n" + "="*80)
print("STAGE 4 — CORRECTED DEEP LEARNING PIPELINE")
print("="*80)

import os
import gc
import json
import numpy as np
import pandas as pd
import tensorflow as tf

from sklearn.preprocessing import StandardScaler
from sklearn.impute import SimpleImputer
from sklearn.metrics import (
    average_precision_score,
    roc_auc_score,
    mean_squared_error,
    r2_score
)

from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Input, Dense, Dropout, LSTM
from tensorflow.keras.callbacks import EarlyStopping
from tensorflow.keras.metrics import AUC


# =============================================================================
# 4.0 — ENVIRONMENT INFORMATION
# =============================================================================

print(f"TensorFlow version: {tf.__version__}")

print("\nAvailable TensorFlow devices:")
print(tf.config.list_physical_devices())

# Force float32 to reduce memory consumption
tf.keras.backend.set_floatx("float32")

print("\nTensorFlow default dtype set to float32.")


# =============================================================================
# 4.1 — VERIFY STAGE 2 INPUTS
# =============================================================================

print("\n" + "-"*80)
print("4.1 VERIFYING STAGE 2 INPUT DATA")
print("-"*80)

# These objects should already have been created by Stage 2
required_objects = [
    "X_train",
    "X_val",
    "X_test",
    "y_train_c",
    "y_val_c",
    "y_test_c",
    "X_train_r",
    "X_val_r",
    "X_test_r",
    "y_train_r",
    "y_val_r",
    "y_test_r"
]

missing_objects = [
    obj for obj in required_objects
    if obj not in globals()
]

if missing_objects:
    raise RuntimeError(
        "Stage 4 cannot continue because the following Stage 2/3 "
        f"objects are missing:\n{missing_objects}"
    )

print("✓ Required Stage 2/3 objects found.")


# =============================================================================
# 4.2 — FEATURE CONTRACT
# =============================================================================

feature_columns = X_train.columns.tolist()

print(f"\nNumber of modelling features: {len(feature_columns)}")

print("\nFeature list:")
for i, feature in enumerate(feature_columns, start=1):
    print(f"{i:02d}. {feature}")

if len(feature_columns) != 21:
    raise ValueError(
        f"Expected 21 modelling features, but found "
        f"{len(feature_columns)}."
    )


# =============================================================================
# 4.3 — MEMORY-SAFE FEATURE PREPARATION
# =============================================================================
#
# IMPORTANT:
#
# We DO NOT create:
#
#     sequence -> flatten -> SimpleImputer
#
# Instead:
#
#     feature matrix
#          ↓
#     train-only median imputation
#          ↓
#     train-only StandardScaler
#          ↓
#     float32
#          ↓
#     sequence generator
#
# This prevents the huge intermediate 19-million-row matrix.
# =============================================================================

print("\n" + "-"*80)
print("4.3 MEMORY-SAFE FEATURE PREPARATION")
print("-"*80)


# Convert to NumPy float32
X_train_np = X_train[feature_columns].to_numpy(dtype=np.float32)
X_val_np   = X_val[feature_columns].to_numpy(dtype=np.float32)
X_test_np  = X_test[feature_columns].to_numpy(dtype=np.float32)

print(f"Raw train matrix: {X_train_np.shape}")
print(f"Raw validation matrix: {X_val_np.shape}")
print(f"Raw test matrix: {X_test_np.shape}")


# -----------------------------------------------------------------------------
# Train-only imputation
# -----------------------------------------------------------------------------

print("\nFitting median imputer on TRAINING data only...")

feature_imputer = SimpleImputer(
    strategy="median"
)

X_train_np = feature_imputer.fit_transform(
    X_train_np
).astype(np.float32)

X_val_np = feature_imputer.transform(
    X_val_np
).astype(np.float32)

X_test_np = feature_imputer.transform(
    X_test_np
).astype(np.float32)


# Verify no NaNs remain
print("\nRemaining missing feature values:")

print(
    f"Train: {np.isnan(X_train_np).sum():,}"
)

print(
    f"Validation: {np.isnan(X_val_np).sum():,}"
)

print(
    f"Test: {np.isnan(X_test_np).sum():,}"
)


# -----------------------------------------------------------------------------
# Train-only scaling
# -----------------------------------------------------------------------------

print("\nFitting StandardScaler on TRAINING data only...")

dl_feature_scaler = StandardScaler()

X_train_np = dl_feature_scaler.fit_transform(
    X_train_np
).astype(np.float32)

X_val_np = dl_feature_scaler.transform(
    X_val_np
).astype(np.float32)

X_test_np = dl_feature_scaler.transform(
    X_test_np
).astype(np.float32)


print("\nFeature scaling complete.")

print(f"Train dtype: {X_train_np.dtype}")
print(f"Validation dtype: {X_val_np.dtype}")
print(f"Test dtype: {X_test_np.dtype}")


# Save preprocessing objects
import joblib

joblib.dump(
    feature_imputer,
    os.path.join(
        output_dir,
        "dl_feature_imputer.pkl"
    )
)

joblib.dump(
    dl_feature_scaler,
    os.path.join(
        output_dir,
        "dl_feature_scaler.pkl"
    )
)

print("\n✓ Deep learning preprocessing objects saved.")


# =============================================================================
# 4.4 — CLASSIFICATION: MLP
# =============================================================================

print("\n" + "="*80)
print("4.4 TASK 1 — DEEP LEARNING CLASSIFICATION")
print("="*80)

print("\nTraining MLP classifier...")


# Convert classification targets to float32
y_train_c_np = np.asarray(
    y_train_c,
    dtype=np.float32
)

y_val_c_np = np.asarray(
    y_val_c,
    dtype=np.float32
)

y_test_c_np = np.asarray(
    y_test_c,
    dtype=np.float32
)


# -----------------------------------------------------------------------------
# MLP architecture
# -----------------------------------------------------------------------------

mlp = Sequential([
    Input(shape=(len(feature_columns),)),

    Dense(
        64,
        activation="relu"
    ),

    Dropout(0.30),

    Dense(
        32,
        activation="relu"
    ),

    Dropout(0.20),

    Dense(
        1,
        activation="sigmoid"
    )
])


mlp.compile(
    optimizer="adam",
    loss="binary_crossentropy",
    metrics=[
        AUC(
            name="roc_auc",
            curve="ROC"
        ),
        AUC(
            name="pr_auc",
            curve="PR"
        )
    ]
)


# -----------------------------------------------------------------------------
# Train MLP
# -----------------------------------------------------------------------------

early_stopping_mlp = EarlyStopping(
    monitor="val_pr_auc",
    mode="max",
    patience=2,
    restore_best_weights=True
)


history_mlp = mlp.fit(
    X_train_np,
    y_train_c_np,

    validation_data=(
        X_val_np,
        y_val_c_np
    ),

    epochs=8,

    batch_size=512,

    callbacks=[
        early_stopping_mlp
    ],

    verbose=1
)


# -----------------------------------------------------------------------------
# MLP Validation prediction
# -----------------------------------------------------------------------------

mlp_val_probs = mlp.predict(
    X_val_np,
    batch_size=4096,
    verbose=0
).flatten()


# -----------------------------------------------------------------------------
# MLP Holdout prediction
# -----------------------------------------------------------------------------

mlp_test_probs = mlp.predict(
    X_test_np,
    batch_size=4096,
    verbose=0
).flatten()


# -----------------------------------------------------------------------------
# MLP metrics
# -----------------------------------------------------------------------------

mlp_val_pr_auc = average_precision_score(
    y_val_c_np,
    mlp_val_probs
)

mlp_val_roc_auc = roc_auc_score(
    y_val_c_np,
    mlp_val_probs
)

mlp_test_pr_auc = average_precision_score(
    y_test_c_np,
    mlp_test_probs
)

mlp_test_roc_auc = roc_auc_score(
    y_test_c_np,
    mlp_test_probs
)


df_dl_class = pd.DataFrame([{
    "Model": "Deep Learning MLP",
    "Task": "Classification",

    "Validation_PR_AUC": round(
        mlp_val_pr_auc,
        6
    ),

    "Validation_ROC_AUC": round(
        mlp_val_roc_auc,
        6
    ),

    "Holdout_PR_AUC": round(
        mlp_test_pr_auc,
        6
    ),

    "Holdout_ROC_AUC": round(
        mlp_test_roc_auc,
        6
    )
}])


print("\n" + "="*80)
print("MLP CLASSIFICATION RESULTS")
print("="*80)

print(
    df_dl_class.to_string(index=False)
)


# Save model
mlp_path = os.path.join(
    output_dir,
    "dl_classifier_mlp.keras"
)

mlp.save(mlp_path)

print(
    f"\n✓ MLP model saved:\n{mlp_path}"
)


# =============================================================================
# 4.5 — PREPARE RUL TARGET
# =============================================================================

print("\n" + "="*80)
print("4.5 TASK 2 — RUL TARGET PREPARATION")
print("="*80)


# Convert RUL targets to float32
y_train_r_np = np.asarray(
    y_train_r,
    dtype=np.float32
)

y_val_r_np = np.asarray(
    y_val_r,
    dtype=np.float32
)

y_test_r_np = np.asarray(
    y_test_r,
    dtype=np.float32
)


print(
    f"Train RUL observations: {len(y_train_r_np):,}"
)

print(
    f"Validation RUL observations: {len(y_val_r_np):,}"
)

print(
    f"Test RUL observations: {len(y_test_r_np):,}"
)


# =============================================================================
# 4.6 — TRAIN-ONLY RUL TARGET SCALING
# =============================================================================

print("\nFitting RUL target scaler on TRAINING data only...")

rul_scaler_dl = StandardScaler()

y_train_r_scaled = rul_scaler_dl.fit_transform(
    y_train_r_np.reshape(-1, 1)
).astype(np.float32).flatten()

y_val_r_scaled = rul_scaler_dl.transform(
    y_val_r_np.reshape(-1, 1)
).astype(np.float32).flatten()

y_test_r_scaled = rul_scaler_dl.transform(
    y_test_r_np.reshape(-1, 1)
).astype(np.float32).flatten()


joblib.dump(
    rul_scaler_dl,
    os.path.join(
        output_dir,
        "dl_rul_target_scaler.pkl"
    )
)

print("✓ RUL target scaler saved.")


# =============================================================================
# 4.7 — LSTM SEQUENCE CONFIGURATION
# =============================================================================

SEQUENCE_LENGTH = 12
SAMPLING_RATE = 1

print("\n" + "-"*80)
print("LSTM SEQUENCE CONFIGURATION")
print("-"*80)

print(
    f"Sequence length: {SEQUENCE_LENGTH} timesteps"
)

print(
    "Sampling interval: 30 minutes"
)

print(
    f"Historical window: approximately "
    f"{SEQUENCE_LENGTH * 30} minutes "
    f"= {SEQUENCE_LENGTH * 0.5:.1f} hours"
)


# =============================================================================
# 4.8 — MEMORY-SAFE SEQUENCE DATASETS
# =============================================================================
#
# CRITICAL DIFFERENCE FROM THE FAILED VERSION:
#
# We do NOT construct:
#
#     X_train_sequences.reshape(...)
#
# followed by:
#
#     SimpleImputer.fit_transform(...)
#
# Instead, features are already imputed/scaled and
# timeseries_dataset_from_array generates windows lazily.
#
# =============================================================================

print("\n" + "="*80)
print("4.8 BUILDING MEMORY-SAFE LSTM DATASETS")
print("="*80)


def create_lstm_dataset(
    X_data,
    y_data,
    sequence_length,
    batch_size=512,
    shuffle=False
):
    """
    Create a memory-efficient TensorFlow time-series dataset.

    Each sample contains:

        X[t-sequence_length+1 : t+1]

    and predicts:

        y[t]

    No full 3D sequence tensor is materialized in memory.
    """

    dataset = tf.keras.utils.timeseries_dataset_from_array(
        data=X_data,
        targets=y_data,
        sequence_length=sequence_length,
        sequence_stride=1,
        sampling_rate=SAMPLING_RATE,
        batch_size=batch_size,
        shuffle=shuffle
    )

    return dataset


LSTM_BATCH_SIZE = 512


train_lstm_ds = create_lstm_dataset(
    X_train_np,
    y_train_r_scaled,
    SEQUENCE_LENGTH,
    batch_size=LSTM_BATCH_SIZE,
    shuffle=True
)


val_lstm_ds = create_lstm_dataset(
    X_val_np,
    y_val_r_scaled,
    SEQUENCE_LENGTH,
    batch_size=LSTM_BATCH_SIZE,
    shuffle=False
)


test_lstm_ds = create_lstm_dataset(
    X_test_np,
    y_test_r_scaled,
    SEQUENCE_LENGTH,
    batch_size=LSTM_BATCH_SIZE,
    shuffle=False
)


print("✓ Training sequence dataset created.")
print("✓ Validation sequence dataset created.")
print("✓ Holdout sequence dataset created.")


# Calculate theoretical sequence counts
train_sequence_count = max(
    0,
    len(X_train_np) - SEQUENCE_LENGTH + 1
)

val_sequence_count = max(
    0,
    len(X_val_np) - SEQUENCE_LENGTH + 1
)

test_sequence_count = max(
    0,
    len(X_test_np) - SEQUENCE_LENGTH + 1
)


print("\nSequence counts:")

print(
    f"Train:      {train_sequence_count:,}"
)

print(
    f"Validation: {val_sequence_count:,}"
)

print(
    f"Test:       {test_sequence_count:,}"
)


# =============================================================================
# 4.9 — BUILD LSTM MODEL
# =============================================================================

print("\n" + "="*80)
print("4.9 BUILDING LSTM REGRESSOR")
print("="*80)


lstm = Sequential([
    Input(
        shape=(
            SEQUENCE_LENGTH,
            len(feature_columns)
        )
    ),

    LSTM(
        64,
        activation="tanh",
        return_sequences=True
    ),

    Dropout(0.20),

    LSTM(
        32,
        activation="tanh"
    ),

    Dropout(0.20),

    Dense(
        1,
        activation="linear"
    )
])


lstm.compile(
    optimizer="adam",
    loss="mse",
    metrics=["mae"]
)


lstm.summary()


# =============================================================================
# 4.10 — TRAIN LSTM
# =============================================================================

print("\n" + "="*80)
print("4.10 TRAINING LSTM RUL REGRESSOR")
print("="*80)


early_stopping_lstm = EarlyStopping(
    monitor="val_loss",
    mode="min",
    patience=2,
    restore_best_weights=True
)


history_lstm = lstm.fit(
    train_lstm_ds,

    validation_data=val_lstm_ds,

    epochs=8,

    callbacks=[
        early_stopping_lstm
    ],

    verbose=1
)


# =============================================================================
# 4.11 — LSTM HOLDOUT PREDICTIONS
# =============================================================================

print("\n" + "="*80)
print("4.11 GENERATING LSTM HOLDOUT PREDICTIONS")
print("="*80)


lstm_test_preds_scaled = lstm.predict(
    test_lstm_ds,
    verbose=1
).flatten()


# -----------------------------------------------------------------------------
# Important:
#
# timeseries_dataset_from_array creates:
#
#     len(X) - sequence_length + 1
#
# samples.
#
# Therefore the corresponding target values are:
#
#     y[sequence_length-1:]
#
# -----------------------------------------------------------------------------

y_test_r_sequence = y_test_r_np[
    SEQUENCE_LENGTH - 1:
]


# Ensure matching lengths
n_test = min(
    len(lstm_test_preds_scaled),
    len(y_test_r_sequence)
)

lstm_test_preds_scaled = (
    lstm_test_preds_scaled[:n_test]
)

y_test_r_sequence = (
    y_test_r_sequence[:n_test]
)


# -----------------------------------------------------------------------------
# Inverse transform predictions
# -----------------------------------------------------------------------------

lstm_test_preds = rul_scaler_dl.inverse_transform(
    lstm_test_preds_scaled.reshape(-1, 1)
).flatten()


# =============================================================================
# 4.12 — LSTM REGRESSION METRICS
# =============================================================================

lstm_rmse = np.sqrt(
    mean_squared_error(
        y_test_r_sequence,
        lstm_test_preds
    )
)

lstm_r2 = r2_score(
    y_test_r_sequence,
    lstm_test_preds
)


df_dl_reg = pd.DataFrame([{
    "Model": "LSTM Regressor",
    "Task": "Regression",

    "Holdout_RMSE_Days": round(
        lstm_rmse,
        6
    ),

    "Holdout_R2": round(
        lstm_r2,
        6
    ),

    "Sequence_Length": SEQUENCE_LENGTH,

    "History_Hours": SEQUENCE_LENGTH * 0.5
}])


print("\n" + "="*80)
print("LSTM REGRESSION RESULTS")
print("="*80)

print(
    df_dl_reg.to_string(index=False)
)


# =============================================================================
# 4.13 — SAVE LSTM MODEL
# =============================================================================

lstm_path = os.path.join(
    output_dir,
    "dl_regressor_lstm.keras"
)

lstm.save(lstm_path)

print(
    f"\n✓ LSTM model saved:\n{lstm_path}"
)


# =============================================================================
# 4.14 — SAVE STAGE 4 METADATA
# =============================================================================

stage4_metadata = {

    "stage": 4,

    "description":
        "Corrected Deep Learning Pipeline",

    "features": feature_columns,

    "feature_count":
        len(feature_columns),

    "classification": {
        "model": "Deep Learning MLP",
        "validation_pr_auc":
            float(mlp_val_pr_auc),
        "validation_roc_auc":
            float(mlp_val_roc_auc),
        "holdout_pr_auc":
            float(mlp_test_pr_auc),
        "holdout_roc_auc":
            float(mlp_test_roc_auc)
    },

    "regression": {
        "model": "LSTM Regressor",
        "sequence_length":
            SEQUENCE_LENGTH,
        "sampling_minutes":
            30,
        "history_hours":
            SEQUENCE_LENGTH * 0.5,
        "holdout_rmse_days":
            float(lstm_rmse),
        "holdout_r2":
            float(lstm_r2)
    },

    "preprocessing": {
        "imputation":
            "Median imputation fitted on training features only",

        "feature_scaling":
            "StandardScaler fitted on training features only",

        "target_scaling":
            "StandardScaler fitted on training RUL only",

        "dtype":
            "float32",

        "sequence_generation":
            "tf.keras.utils.timeseries_dataset_from_array"
    },

    "memory_correction":
        "Avoided flattening all LSTM sequences and applying "
        "SimpleImputer to the resulting multi-million-row matrix."
}


metadata_path = os.path.join(
    output_dir,
    "stage4_deep_learning_metadata.json"
)

with open(
    metadata_path,
    "w"
) as f:

    json.dump(
        stage4_metadata,
        f,
        indent=4
    )


# =============================================================================
# 4.15 — FINAL STAGE 4 SUMMARY
# =============================================================================

print("\n" + "="*80)
print("STAGE 4 COMPLETE")
print("="*80)

print("\nCLASSIFICATION — MLP")
print(
    f"Validation PR-AUC : {mlp_val_pr_auc:.6f}"
)
print(
    f"Validation ROC-AUC: {mlp_val_roc_auc:.6f}"
)
print(
    f"Holdout PR-AUC    : {mlp_test_pr_auc:.6f}"
)
print(
    f"Holdout ROC-AUC   : {mlp_test_roc_auc:.6f}"
)

print("\nREGRESSION — LSTM")
print(
    f"Holdout RMSE      : {lstm_rmse:.6f} days"
)
print(
    f"Holdout R²        : {lstm_r2:.6f}"
)

print("\nSequence configuration:")
print(
    f"Length            : {SEQUENCE_LENGTH} timesteps"
)
print(
    f"History           : {SEQUENCE_LENGTH * 0.5:.1f} hours"
)

print("\nMemory-safe implementation:")
print(
    "✓ Train-only feature imputation"
)
print(
    "✓ Train-only feature scaling"
)
print(
    "✓ float32 tensors"
)
print(
    "✓ Lazy sequence generation"
)
print(
    "✓ No flattened multi-million-row sequence matrix"
)
print(
    "✓ No median imputation across sequence windows"
)

print("\n" + "="*80)
```

```python
# ============================================================
# STAGE 5 — COMPUTE SHAP FOR RANDOM FOREST
# ============================================================

import os
import json
import numpy as np
import pandas as pd
import shap
import matplotlib.pyplot as plt

print("=" * 80)
print("STAGE 5 — RANDOM FOREST SHAP EXPLAINABILITY")
print("=" * 80)


# ------------------------------------------------------------
# 1. PATHS
# ------------------------------------------------------------

stage2_dir = r"D:\Data_Workspace\Capstone_Project\NLGN_PLC_Predictive_Analystic\data\04_corrected_pipeline\stage_2_cleaning"

stage5_dir = r"D:\Data_Workspace\Capstone_Project\NLGN_PLC_Predictive_Analystic\data\04_corrected_pipeline\stage_5_explainability"

os.makedirs(stage5_dir, exist_ok=True)


# ------------------------------------------------------------
# 2. LOAD STAGE 2 CLEANED DATA
# ------------------------------------------------------------

cleaned_path = os.path.join(
    stage2_dir,
    "NLNG_cleaned_leakage_controlled.csv"
)

df_stage2 = pd.read_csv(cleaned_path)

print(f"Rows loaded: {len(df_stage2):,}")


# ------------------------------------------------------------
# 3. AUTHORITATIVE 21-FEATURE CONTRACT
# ------------------------------------------------------------

feature_columns = [
    "commission_year",
    "asset_age_years",
    "hours_since_maint",
    "cumulative_op_hours",
    "bearing_temperature",
    "rpm",
    "vibration",
    "overall_vibration",
    "motor_current",
    "oil_pressure",
    "oil_particles_ppm",
    "bearing_index",
    "discharge_pressure",
    "feed_gas_pressure",
    "lng_output_tph",
    "ambient_temperature",
    "load_factor",
    "wear_level",
    "lubrication_health_index",
    "production_efficiency",
    "quality_factor"
]

X_raw = df_stage2[feature_columns].copy()

print(f"Feature count: {X_raw.shape[1]}")


# ------------------------------------------------------------
# 4. CREATE SMALL SHAP SAMPLE
# ------------------------------------------------------------

SAMPLE_SIZE = min(
    5000,
    len(X_raw)
)

X_sample = X_raw.sample(
    n=SAMPLE_SIZE,
    random_state=42
).reset_index(drop=True)

print(
    f"SHAP sample size: {len(X_sample):,}"
)


# ------------------------------------------------------------
# 5. EXTRACT THE FITTED IMPUTER FROM THE RF PIPELINE
# ------------------------------------------------------------

rf_imputer = rf_classifier.named_steps["imputer"]

rf_model = rf_classifier.named_steps["model"]

print(
    "\nRandom Forest estimator:",
    type(rf_model).__name__
)

print(
    "Number of trees:",
    rf_model.n_estimators
)


# ------------------------------------------------------------
# 6. APPLY THE EXACT FITTED IMPUTER
# ------------------------------------------------------------
#
# IMPORTANT:
# We use the imputer that was already FIT during Stage 3.
#
# We DO NOT fit another imputer here.
# ------------------------------------------------------------

X_sample_imputed = rf_imputer.transform(
    X_sample
)

print(
    "\nTransformed SHAP matrix:",
    X_sample_imputed.shape
)


# ------------------------------------------------------------
# 7. GET FEATURE NAMES AFTER add_indicator=True
# ------------------------------------------------------------

try:

    transformed_feature_names = (
        rf_imputer.get_feature_names_out(
            feature_columns
        )
    )

except Exception:

    transformed_feature_names = np.array(
        feature_columns
    )

print(
    f"Transformed feature count: "
    f"{len(transformed_feature_names)}"
)


# ------------------------------------------------------------
# 8. COMPUTE TREE SHAP
# ------------------------------------------------------------

print("\nCalculating SHAP values...")

explainer = shap.TreeExplainer(
    rf_model
)

shap_output = explainer(
    X_sample_imputed
)

print("✓ SHAP calculation completed.")


# ------------------------------------------------------------
# 9. NORMALISE SHAP OUTPUT
# ------------------------------------------------------------

if hasattr(
    shap_output,
    "values"
):

    shap_values = shap_output.values

else:

    shap_values = np.asarray(
        shap_output
    )


print(
    "Raw SHAP shape:",
    shap_values.shape
)


# ------------------------------------------------------------
# 10. SELECT FAILURE CLASS
# ------------------------------------------------------------

if shap_values.ndim == 3:

    # Expected:
    # samples × features × classes

    if shap_values.shape[2] >= 2:

        shap_failure = shap_values[:, :, 1]

    else:

        shap_failure = shap_values[:, :, 0]

elif shap_values.ndim == 2:

    shap_failure = shap_values

else:

    raise ValueError(
        f"Unexpected SHAP shape: "
        f"{shap_values.shape}"
    )


print(
    "Failure-class SHAP shape:",
    shap_failure.shape
)


# ------------------------------------------------------------
# 11. GLOBAL SHAP IMPORTANCE
# ------------------------------------------------------------

mean_abs_shap = (
    np.abs(shap_failure)
    .mean(axis=0)
)

shap_importance = pd.DataFrame({
    "Feature": transformed_feature_names,
    "Mean_Absolute_SHAP": mean_abs_shap
})

shap_importance = (
    shap_importance
    .sort_values(
        "Mean_Absolute_SHAP",
        ascending=False
    )
    .reset_index(drop=True)
)


# ------------------------------------------------------------
# 12. DISPLAY TOP 15
# ------------------------------------------------------------

print("\n" + "=" * 80)
print("TOP 15 SHAP FEATURES")
print("=" * 80)

display(
    shap_importance.head(15)
)


# ------------------------------------------------------------
# 13. SAVE SHAP CSV
# ------------------------------------------------------------

shap_csv_path = os.path.join(
    stage5_dir,
    "shap_feature_importance.csv"
)

shap_importance.to_csv(
    shap_csv_path,
    index=False
)

print(
    f"\n✓ SHAP table saved:\n"
    f"{shap_csv_path}"
)


# ------------------------------------------------------------
# 14. TOP-10 SHAP BAR CHART
# ------------------------------------------------------------

top_n = min(
    10,
    len(shap_importance)
)

plot_df = (
    shap_importance
    .head(top_n)
    .sort_values(
        "Mean_Absolute_SHAP"
    )
)

plt.figure(
    figsize=(10, 6)
)

plt.barh(
    plot_df["Feature"],
    plot_df["Mean_Absolute_SHAP"]
)

plt.xlabel(
    "Mean Absolute SHAP Value"
)

plt.ylabel(
    "Feature"
)

plt.title(
    "Top 10 Features Influencing 24-Hour Failure Prediction"
)

plt.tight_layout()

shap_plot_path = os.path.join(
    stage5_dir,
    "shap_feature_importance.png"
)

plt.savefig(
    shap_plot_path,
    dpi=300,
    bbox_inches="tight"
)

plt.show()

plt.close()


# ------------------------------------------------------------
# 15. SAVE METADATA
# ------------------------------------------------------------

stage5_metadata = {

    "stage": 5,

    "model": "Random Forest Classifier",

    "target":
        "failure_within_24h",

    "shap_method":
        "TreeExplainer",

    "sample_size":
        int(SAMPLE_SIZE),

    "original_feature_count":
        len(feature_columns),

    "transformed_feature_count":
        len(transformed_feature_names),

    "imputer":
        "Stage 3 fitted SimpleImputer",

    "imputation_strategy":
        "median",

    "missing_indicators":
        True,

    "outputs": {
        "importance_csv":
            "shap_feature_importance.csv",

        "importance_plot":
            "shap_feature_importance.png"
    }
}

metadata_path = os.path.join(
    stage5_dir,
    "stage5_shap_metadata.json"
)

with open(
    metadata_path,
    "w",
    encoding="utf-8"
) as f:

    json.dump(
        stage5_metadata,
        f,
        indent=4
    )

print(
    f"\n✓ Metadata saved:\n"
    f"{metadata_path}"
)


# ------------------------------------------------------------
# 16. FINAL CHECK
# ------------------------------------------------------------

print("\n" + "=" * 80)
print("STAGE 5 COMPLETE")
print("=" * 80)

print(
    f"Model: Random Forest Classifier"
)

print(
    f"SHAP sample: {SAMPLE_SIZE:,}"
)

print(
    f"Original features: {len(feature_columns)}"
)

print(
    f"Explained features: {len(transformed_feature_names)}"
)

print(
    "\nOutputs:"
)

print(
    f"✓ {shap_csv_path}"
)

print(
    f"✓ {shap_plot_path}"
)

print(
    f"✓ {metadata_path}"
)

print("=" * 80)
```

```python
# =============================================================================
# IDSS MODEL-SERVING LAYER — STEP 3
# Create the actual model service
# =============================================================================

import os
import textwrap

# -------------------------------------------------------------------------
# PATHS
# -------------------------------------------------------------------------

BACKEND_DIR = os.path.join(
    CORRECTED_DIR,
    "idss_backend"
)

SERVICES_DIR = os.path.join(
    BACKEND_DIR,
    "services"
)

SCHEMAS_DIR = os.path.join(
    BACKEND_DIR,
    "schemas"
)

os.makedirs(SERVICES_DIR, exist_ok=True)
os.makedirs(SCHEMAS_DIR, exist_ok=True)


# =============================================================================
# 1. AUTHORITATIVE FEATURE CONTRACT
# =============================================================================

FEATURE_COLUMNS = [
    "commission_year",
    "asset_age_years",
    "hours_since_maint",
    "cumulative_op_hours",
    "bearing_temperature",
    "rpm",
    "vibration",
    "overall_vibration",
    "motor_current",
    "oil_pressure",
    "oil_particles_ppm",
    "bearing_index",
    "discharge_pressure",
    "feed_gas_pressure",
    "lng_output_tph",
    "ambient_temperature",
    "load_factor",
    "wear_level",
    "lubrication_health_index",
    "production_efficiency",
    "quality_factor"
]


# =============================================================================
# 2. MODEL PATHS
# =============================================================================

MLP_MODEL_PATH = os.path.join(
    STAGE4_DIR,
    "dl_classifier_mlp.keras"
)

MLP_IMPUTER_PATH = os.path.join(
    STAGE4_DIR,
    "dl_feature_imputer.pkl"
)

MLP_SCALER_PATH = os.path.join(
    STAGE4_DIR,
    "dl_feature_scaler.pkl"
)

GB_REGRESSOR_PATH = os.path.join(
    STAGE3_DIR,
    "gradient_boosting_regressor.pkl"
)


# =============================================================================
# 3. VERIFY REQUIRED ARTEFACTS
# =============================================================================

required_artifacts = {
    "MLP classifier": MLP_MODEL_PATH,
    "MLP feature imputer": MLP_IMPUTER_PATH,
    "MLP feature scaler": MLP_SCALER_PATH,
    "Gradient Boosting regressor": GB_REGRESSOR_PATH
}

print("=" * 80)
print("STEP 3 — VERIFYING IDSS MODEL ARTEFACTS")
print("=" * 80)

missing = []

for name, path in required_artifacts.items():

    if os.path.exists(path):

        size_mb = os.path.getsize(path) / (1024 ** 2)

        print(
            f"✓ {name:<30} "
            f"{os.path.basename(path):<40} "
            f"{size_mb:.2f} MB"
        )

    else:

        print(
            f"✗ MISSING: {name}\n"
            f"  {path}"
        )

        missing.append(path)

if missing:

    raise FileNotFoundError(
        "Required IDSS model artefacts are missing."
    )

print("\n✓ All required model artefacts are available.")


# =============================================================================
# 4. CREATE model_service.py
# =============================================================================

model_service_code = r'''
import os
import joblib
import numpy as np
import pandas as pd
import tensorflow as tf


# ============================================================================
# MODEL PATHS
# ============================================================================

BASE_DIR = os.path.dirname(
    os.path.dirname(
        os.path.abspath(__file__)
    )
)

CORRECTED_DIR = os.path.dirname(
    BASE_DIR
)

STAGE3_DIR = os.path.join(
    CORRECTED_DIR,
    "stage_3_ml"
)

STAGE4_DIR = os.path.join(
    CORRECTED_DIR,
    "stage_4_deep_learning"
)


MLP_MODEL_PATH = os.path.join(
    STAGE4_DIR,
    "dl_classifier_mlp.keras"
)

MLP_IMPUTER_PATH = os.path.join(
    STAGE4_DIR,
    "dl_feature_imputer.pkl"
)

MLP_SCALER_PATH = os.path.join(
    STAGE4_DIR,
    "dl_feature_scaler.pkl"
)

GB_REGRESSOR_PATH = os.path.join(
    STAGE3_DIR,
    "gradient_boosting_regressor.pkl"
)


# ============================================================================
# AUTHORITATIVE FEATURE CONTRACT
# ============================================================================

FEATURE_COLUMNS = [
    "commission_year",
    "asset_age_years",
    "hours_since_maint",
    "cumulative_op_hours",
    "bearing_temperature",
    "rpm",
    "vibration",
    "overall_vibration",
    "motor_current",
    "oil_pressure",
    "oil_particles_ppm",
    "bearing_index",
    "discharge_pressure",
    "feed_gas_pressure",
    "lng_output_tph",
    "ambient_temperature",
    "load_factor",
    "wear_level",
    "lubrication_health_index",
    "production_efficiency",
    "quality_factor"
]


# ============================================================================
# MODEL SERVICE
# ============================================================================

class ModelService:

    def __init__(self):

        self.classifier = None
        self.feature_imputer = None
        self.feature_scaler = None
        self.regressor = None

        self.loaded = False

        self.load_models()


    # ------------------------------------------------------------------------
    # LOAD MODELS
    # ------------------------------------------------------------------------

    def load_models(self):

        # Load MLP classifier
        self.classifier = tf.keras.models.load_model(
            MLP_MODEL_PATH,
            compile=False
        )

        # Load preprocessing
        self.feature_imputer = joblib.load(
            MLP_IMPUTER_PATH
        )

        self.feature_scaler = joblib.load(
            MLP_SCALER_PATH
        )

        # Load Gradient Boosting RUL model
        self.regressor = joblib.load(
            GB_REGRESSOR_PATH
        )

        self.loaded = True


    # ------------------------------------------------------------------------
    # VALIDATE INPUT
    # ------------------------------------------------------------------------

    def validate_features(
        self,
        data: dict
    ):

        missing = [
            feature
            for feature in FEATURE_COLUMNS
            if feature not in data
        ]

        if missing:

            raise ValueError(
                "Missing required features: "
                + ", ".join(missing)
            )


    # ------------------------------------------------------------------------
    # PREPARE INPUT FOR MLP
    # ------------------------------------------------------------------------

    def prepare_classifier_input(
        self,
        data: dict
    ):

        self.validate_features(data)

        X = pd.DataFrame(
            [[
                data[feature]
                for feature in FEATURE_COLUMNS
            ]],
            columns=FEATURE_COLUMNS
        )

        # Apply exact Stage 4 training preprocessing
        X_imputed = self.feature_imputer.transform(
            X
        )

        X_scaled = self.feature_scaler.transform(
            X_imputed
        )

        return X_scaled.astype(
            np.float32
        )


    # ------------------------------------------------------------------------
    # FAILURE PREDICTION
    # ------------------------------------------------------------------------

    def predict_failure(
        self,
        data: dict
    ):

        X_scaled = self.prepare_classifier_input(
            data
        )

        probability = float(
            self.classifier.predict(
                X_scaled,
                verbose=0
            ).reshape(-1)[0]
        )

        probability = max(
            0.0,
            min(
                1.0,
                probability
            )
        )

        if probability >= 0.70:

            risk_level = "CRITICAL"

        elif probability >= 0.30:

            risk_level = "HIGH"

        elif probability >= 0.10:

            risk_level = "MODERATE"

        else:

            risk_level = "LOW"

        return {
            "failure_probability_24h": probability,
            "failure_risk": risk_level
        }


    # ------------------------------------------------------------------------
    # RUL PREDICTION
    # ------------------------------------------------------------------------

    def predict_rul(
        self,
        data: dict
    ):

        self.validate_features(data)

        X = pd.DataFrame(
            [[
                data[feature]
                for feature in FEATURE_COLUMNS
            ]],
            columns=FEATURE_COLUMNS
        )

        prediction = float(
            np.asarray(
                self.regressor.predict(X)
            ).reshape(-1)[0]
        )

        # Prevent impossible negative display values
        prediction = max(
            0.0,
            prediction
        )

        return {
            "rul_days": prediction
        }


    # ------------------------------------------------------------------------
    # COMPLETE PREDICTION
    # ------------------------------------------------------------------------

    def predict(
        self,
        data: dict
    ):

        failure_result = self.predict_failure(
            data
        )

        rul_result = self.predict_rul(
            data
        )

        return {
            **failure_result,
            **rul_result,
            "classification_model":
                "Deep Learning MLP",

            "regression_model":
                "Gradient Boosting Regressor"
        }


# ============================================================================
# SINGLE SHARED SERVICE INSTANCE
# ============================================================================

model_service = ModelService()
'''


model_service_path = os.path.join(
    SERVICES_DIR,
    "model_service.py"
)

with open(
    model_service_path,
    "w",
    encoding="utf-8"
) as f:

    f.write(model_service_code)


# =============================================================================
# 5. CREATE schemas.py
# =============================================================================

schemas_code = r'''
from typing import Optional

from pydantic import BaseModel, Field


class PredictionRequest(BaseModel):

    equipment_id: Optional[str] = None

    train: Optional[str] = None

    commission_year: float
    asset_age_years: float
    hours_since_maint: float
    cumulative_op_hours: float
    bearing_temperature: float
    rpm: float
    vibration: float
    overall_vibration: float
    motor_current: float
    oil_pressure: float
    oil_particles_ppm: float
    bearing_index: float
    discharge_pressure: float
    feed_gas_pressure: float
    lng_output_tph: float
    ambient_temperature: float
    load_factor: float
    wear_level: float
    lubrication_health_index: float
    production_efficiency: float
    quality_factor: float


class PredictionResponse(BaseModel):

    equipment_id: Optional[str] = None

    train: Optional[str] = None

    failure_probability_24h: float = Field(
        ge=0.0,
        le=1.0
    )

    failure_risk: str

    rul_days: float

    classification_model: str

    regression_model: str


class HealthResponse(BaseModel):

    status: str

    classification_model: str

    regression_model: str

    feature_count: int
'''


schemas_path = os.path.join(
    SCHEMAS_DIR,
    "schemas.py"
)

with open(
    schemas_path,
    "w",
    encoding="utf-8"
) as f:

    f.write(schemas_code)


# =============================================================================
# 6. CREATE __init__.py FILES
# =============================================================================

for directory in [
    BACKEND_DIR,
    SERVICES_DIR,
    SCHEMAS_DIR
]:

    init_path = os.path.join(
        directory,
        "__init__.py"
    )

    if not os.path.exists(init_path):

        with open(
            init_path,
            "w",
            encoding="utf-8"
        ) as f:

            f.write("")


# =============================================================================
# FINAL OUTPUT
# =============================================================================

print("\n" + "=" * 80)
print("STEP 3 COMPLETE")
print("=" * 80)

print(
    f"\nCreated:\n"
    f"✓ {model_service_path}\n"
    f"✓ {schemas_path}"
)

print(
    "\nAuthoritative feature count:",
    len(FEATURE_COLUMNS)
)

print("\nModels:")
print("✓ Deep Learning MLP classifier")
print("✓ Gradient Boosting Regressor")

print("\nNext step:")
print("We will test the model service directly before starting FastAPI.")
```

```python
# =============================================================================
# IDSS MODEL-SERVING LAYER — STEP 4
# DIRECT TEST OF THE ACTUAL PIPELINE 2.0 MODELS
# =============================================================================

import os
import sys
import pandas as pd
import numpy as np

print("=" * 80)
print("STEP 4 — DIRECT MODEL SERVICE TEST")
print("=" * 80)


# -----------------------------------------------------------------------------
# 1. BACKEND PATH
# -----------------------------------------------------------------------------

BACKEND_DIR = os.path.join(
    CORRECTED_DIR,
    "idss_backend"
)

SERVICES_DIR = os.path.join(
    BACKEND_DIR,
    "services"
)

# Make backend importable
if BACKEND_DIR not in sys.path:
    sys.path.insert(
        0,
        BACKEND_DIR
    )


# -----------------------------------------------------------------------------
# 2. LOAD THE MODEL SERVICE
# -----------------------------------------------------------------------------

print("\nLoading ModelService...")

from services.model_service import (
    model_service,
    FEATURE_COLUMNS
)

print("✓ Model service imported.")
print("✓ Models loaded successfully.")


# -----------------------------------------------------------------------------
# 3. VERIFY MODEL SERVICE STATE
# -----------------------------------------------------------------------------

print("\n" + "-" * 80)
print("MODEL SERVICE STATUS")
print("-" * 80)

print(
    "Service loaded:",
    model_service.loaded
)

print(
    "Classification model:",
    type(model_service.classifier).__name__
)

print(
    "Regression model:",
    type(model_service.regressor).__name__
)

print(
    "Feature count:",
    len(FEATURE_COLUMNS)
)


# -----------------------------------------------------------------------------
# 4. LOAD A REAL STAGE 2 ROW
# -----------------------------------------------------------------------------

stage2_path = os.path.join(
    CORRECTED_DIR,
    "stage_2_cleaning",
    "NLNG_cleaned_leakage_controlled.csv"
)

print("\n" + "-" * 80)
print("LOADING REAL STAGE 2 DATA")
print("-" * 80)

if not os.path.exists(stage2_path):

    raise FileNotFoundError(
        "Stage 2 cleaned dataset not found:\n"
        + stage2_path
    )

# Read one row only
sample_df = pd.read_csv(
    stage2_path,
    nrows=1
)

print(
    "✓ Stage 2 dataset loaded."
)

print(
    "Sample shape:",
    sample_df.shape
)


# -----------------------------------------------------------------------------
# 5. BUILD PREDICTION INPUT
# -----------------------------------------------------------------------------

sample_row = sample_df.iloc[0]

prediction_input = {}

for feature in FEATURE_COLUMNS:

    value = sample_row[feature]

    # Convert pandas/numpy numeric values to
    # normal Python float for the API/service layer.
    if pd.isna(value):
        prediction_input[feature] = np.nan
    else:
        prediction_input[feature] = float(value)


# -----------------------------------------------------------------------------
# 6. PRESERVE EQUIPMENT METADATA
# -----------------------------------------------------------------------------

equipment_id = (
    sample_row["equipment_id"]
    if "equipment_id" in sample_row.index
    else None
)

train = (
    sample_row["train"]
    if "train" in sample_row.index
    else None
)

timestamp = (
    sample_row["timestamp"]
    if "timestamp" in sample_row.index
    else None
)


print("\n" + "-" * 80)
print("TEST EQUIPMENT")
print("-" * 80)

print(
    "Equipment ID:",
    equipment_id
)

print(
    "Train:",
    train
)

print(
    "Timestamp:",
    timestamp
)


# -----------------------------------------------------------------------------
# 7. RUN THE ACTUAL MODEL SERVICE
# -----------------------------------------------------------------------------

print("\n" + "-" * 80)
print("RUNNING ACTUAL PIPELINE 2.0 MODELS")
print("-" * 80)

try:

    prediction = model_service.predict(
        prediction_input
    )

except Exception as e:

    print("\n✗ MODEL SERVICE FAILED")
    print(
        "\nError type:",
        type(e).__name__
    )

    print(
        "\nError message:",
        str(e)
    )

    raise


# -----------------------------------------------------------------------------
# 8. DISPLAY RESULTS
# -----------------------------------------------------------------------------

print("\n" + "=" * 80)
print("MODEL SERVICE PREDICTION")
print("=" * 80)

print(
    f"Equipment ID: {equipment_id}"
)

print(
    f"Train: {train}"
)

print(
    f"Timestamp: {timestamp}"
)

print(
    "\n24-Hour Failure Probability:",
    f"{prediction['failure_probability_24h']:.6f}"
)

print(
    "24-Hour Failure Probability (%):",
    f"{prediction['failure_probability_24h'] * 100:.2f}%"
)

print(
    "Failure Risk Level:",
    prediction["failure_risk"]
)

print(
    "\nPredicted RUL:",
    f"{prediction['rul_days']:.4f} days"
)

print(
    "\nClassification Model:",
    prediction["classification_model"]
)

print(
    "Regression Model:",
    prediction["regression_model"]
)


# -----------------------------------------------------------------------------
# 9. FINAL VALIDATION
# -----------------------------------------------------------------------------

failure_probability = (
    prediction[
        "failure_probability_24h"
    ]
)

rul_days = (
    prediction[
        "rul_days"
    ]
)

checks = {

    "Failure probability is numeric":
        isinstance(
            failure_probability,
            (float, int)
        ),

    "Failure probability within 0-1":
        0.0 <= failure_probability <= 1.0,

    "RUL is numeric":
        isinstance(
            rul_days,
            (float, int)
        ),

    "RUL is non-negative":
        rul_days >= 0,

    "Correct classifier":
        prediction[
            "classification_model"
        ] == "Deep Learning MLP",

    "Correct regressor":
        prediction[
            "regression_model"
        ] == "Gradient Boosting Regressor"
}


print("\n" + "-" * 80)
print("VALIDATION CHECKS")
print("-" * 80)

all_passed = True

for name, passed in checks.items():

    if passed:
        print(
            f"✓ {name}"
        )

    else:
        print(
            f"✗ {name}"
        )
        all_passed = False


if not all_passed:

    raise RuntimeError(
        "One or more model-service validation checks failed."
    )


# -----------------------------------------------------------------------------
# 10. SAVE TEST RESULT
# -----------------------------------------------------------------------------

test_result = {

    "equipment_id":
        equipment_id,

    "train":
        train,

    "timestamp":
        timestamp,

    "failure_probability_24h":
        failure_probability,

    "failure_risk":
        prediction[
            "failure_risk"
        ],

    "rul_days":
        rul_days,

    "classification_model":
        prediction[
            "classification_model"
        ],

    "regression_model":
        prediction[
            "regression_model"
        ]
}

test_result_df = pd.DataFrame(
    [test_result]
)

test_output_path = os.path.join(
    BACKEND_DIR,
    "model_service_test_result.csv"
)

test_result_df.to_csv(
    test_output_path,
    index=False
)


print("\n" + "=" * 80)
print("STEP 4 COMPLETE")
print("=" * 80)

print(
    "\n✓ Actual Stage 2 data successfully passed "
    "through the actual Pipeline 2.0 models."
)

print(
    f"\nTest result saved to:\n"
    f"{test_output_path}"
)

print("=" * 80)
```

```python
# =============================================================================
# IDSS MODEL-SERVING LAYER — STEP 5
# CREATE FASTAPI APPLICATION
# =============================================================================

import os

BACKEND_DIR = os.path.join(
    CORRECTED_DIR,
    "idss_backend"
)

APP_PATH = os.path.join(
    BACKEND_DIR,
    "app.py"
)

REQUIREMENTS_PATH = os.path.join(
    BACKEND_DIR,
    "requirements.txt"
)

README_PATH = os.path.join(
    BACKEND_DIR,
    "README.md"
)

print("=" * 80)
print("STEP 5 — CREATING FASTAPI MODEL-SERVING API")
print("=" * 80)


# =============================================================================
# 1. FASTAPI APPLICATION
# =============================================================================

app_code = '''
from typing import List

from fastapi import FastAPI, HTTPException

from schemas.schemas import (
    PredictionRequest,
    PredictionResponse,
    HealthResponse
)

from services.model_service import (
    model_service,
    FEATURE_COLUMNS
)


app = FastAPI(
    title="NLNG Predictive Maintenance IDSS API",
    description=(
        "Model-serving API for the NLNG predictive maintenance "
        "decision-support system."
    ),
    version="1.0.0"
)


@app.get("/")
def root():
    return {
        "application": "NLNG Predictive Maintenance IDSS",
        "status": "running",
        "classification_model": "Deep Learning MLP",
        "regression_model": "Gradient Boosting Regressor",
        "feature_count": len(FEATURE_COLUMNS)
    }


@app.get(
    "/health",
    response_model=HealthResponse
)
def health_check():

    if not model_service.loaded:
        raise HTTPException(
            status_code=503,
            detail="Model service is not ready."
        )

    return {
        "status": "healthy",
        "classification_model": "Deep Learning MLP",
        "regression_model": "Gradient Boosting Regressor",
        "feature_count": len(FEATURE_COLUMNS)
    }


@app.get("/model-info")
def model_info():

    return {
        "classification": {
            "model": "Deep Learning MLP",
            "task": "24-hour failure classification",
            "output": "failure probability"
        },
        "regression": {
            "model": "Gradient Boosting Regressor",
            "task": "Remaining Useful Life regression",
            "output": "rul_days"
        },
        "feature_count": len(FEATURE_COLUMNS),
        "features": FEATURE_COLUMNS
    }


@app.post(
    "/predict",
    response_model=PredictionResponse
)
def predict(request: PredictionRequest):

    try:
        input_data = request.model_dump()

        equipment_id = input_data.pop(
            "equipment_id",
            None
        )

        train = input_data.pop(
            "train",
            None
        )

        result = model_service.predict(
            input_data
        )

        return {
            "equipment_id": equipment_id,
            "train": train,
            "failure_probability_24h":
                result["failure_probability_24h"],
            "failure_risk":
                result["failure_risk"],
            "rul_days":
                result["rul_days"],
            "classification_model":
                result["classification_model"],
            "regression_model":
                result["regression_model"]
        }

    except ValueError as exc:

        raise HTTPException(
            status_code=422,
            detail=str(exc)
        )

    except Exception as exc:

        raise HTTPException(
            status_code=500,
            detail="Prediction service error: " + str(exc)
        )


@app.post("/predict/batch")
def predict_batch(
    requests: List[PredictionRequest]
):

    if not requests:

        raise HTTPException(
            status_code=400,
            detail="No prediction records supplied."
        )

    results = []

    for request in requests:

        try:
            input_data = request.model_dump()

            equipment_id = input_data.pop(
                "equipment_id",
                None
            )

            train = input_data.pop(
                "train",
                None
            )

            result = model_service.predict(
                input_data
            )

            results.append({
                "equipment_id": equipment_id,
                "train": train,
                "failure_probability_24h":
                    result["failure_probability_24h"],
                "failure_risk":
                    result["failure_risk"],
                "rul_days":
                    result["rul_days"],
                "classification_model":
                    result["classification_model"],
                "regression_model":
                    result["regression_model"]
            })

        except Exception as exc:

            raise HTTPException(
                status_code=422,
                detail="Batch record failed: " + str(exc)
            )

    return {
        "count": len(results),
        "predictions": results
    }
'''

with open(
    APP_PATH,
    "w",
    encoding="utf-8"
) as f:
    f.write(app_code)

print("✓ app.py created")


# =============================================================================
# 2. REQUIREMENTS
# =============================================================================

requirements_code = """fastapi
uvicorn
pydantic
pandas
numpy
scikit-learn
joblib
tensorflow
"""

with open(
    REQUIREMENTS_PATH,
    "w",
    encoding="utf-8"
) as f:
    f.write(requirements_code)

print("✓ requirements.txt created")


# =============================================================================
# 3. README
# =============================================================================

readme_code = """# NLNG Predictive Maintenance IDSS - FastAPI

## Models

Classification:
Deep Learning MLP

Regression:
Gradient Boosting Regressor

## Endpoints

GET /
GET /health
GET /model-info
POST /predict
POST /predict/batch

## Start

python -m uvicorn app:app --host 127.0.0.1 --port 8000

## Documentation

http://127.0.0.1:8000/docs
"""

with open(
    README_PATH,
    "w",
    encoding="utf-8"
) as f:
    f.write(readme_code)

print("✓ README.md created")


# =============================================================================
# 4. VERIFY
# =============================================================================

print("\n" + "-" * 80)
print("VERIFYING FASTAPI FILES")
print("-" * 80)

for path in [
    APP_PATH,
    REQUIREMENTS_PATH,
    README_PATH
]:

    if os.path.exists(path):
        print(
            f"✓ {os.path.basename(path)}"
        )
    else:
        print(
            f"✗ Missing: {path}"
        )


print("\n" + "=" * 80)
print("STEP 5 COMPLETE")
print("=" * 80)

print(
    "\nFastAPI application files are ready."
)

print(
    "\nBackend directory:"
)

print(
    BACKEND_DIR
)

print(
    "\nNext step: start the FastAPI server."
)

print("=" * 80)
```

```python
# =============================================================================
# CREATE FINAL FASTAPI BACKEND
# Preserve existing Streamlit prototype
# =============================================================================

import os

PROJECT_ROOT = (
    r"D:\Data_Workspace\Capstone_Project"
    r"\NLGN_PLC_Predictive_Analystic"
)

IDSS_DIR = os.path.join(
    PROJECT_ROOT,
    "idss"
)

BACKEND_DIR = os.path.join(
    IDSS_DIR,
    "backend"
)

SERVICES_DIR = os.path.join(
    BACKEND_DIR,
    "services"
)

SCHEMAS_DIR = os.path.join(
    BACKEND_DIR,
    "schemas"
)

os.makedirs(SERVICES_DIR, exist_ok=True)
os.makedirs(SCHEMAS_DIR, exist_ok=True)

# -------------------------------------------------------------------------
# FASTAPI app.py
# -------------------------------------------------------------------------

fastapi_app = '''
from typing import List

from fastapi import FastAPI, HTTPException

from schemas.schemas import (
    PredictionRequest,
    PredictionResponse,
    HealthResponse
)

from services.model_service import (
    model_service,
    FEATURE_COLUMNS
)


app = FastAPI(
    title="NLNG Predictive Maintenance IDSS API",
    description=(
        "REST model-serving API for the NLNG predictive maintenance "
        "decision-support system."
    ),
    version="1.0.0"
)


@app.get("/")
def root():
    return {
        "application": "NLNG Predictive Maintenance IDSS",
        "status": "running",
        "classification_model": "Deep Learning MLP",
        "regression_model": "Gradient Boosting Regressor",
        "feature_count": len(FEATURE_COLUMNS)
    }


@app.get(
    "/health",
    response_model=HealthResponse
)
def health_check():

    if not model_service.loaded:
        raise HTTPException(
            status_code=503,
            detail="Model service is not ready."
        )

    return {
        "status": "healthy",
        "classification_model": "Deep Learning MLP",
        "regression_model": "Gradient Boosting Regressor",
        "feature_count": len(FEATURE_COLUMNS)
    }


@app.get("/model-info")
def model_info():

    return {
        "classification": {
            "model": "Deep Learning MLP",
            "task": "24-hour failure classification",
            "output": "failure probability"
        },
        "regression": {
            "model": "Gradient Boosting Regressor",
            "task": "Remaining Useful Life regression",
            "output": "rul_days"
        },
        "feature_count": len(FEATURE_COLUMNS),
        "features": FEATURE_COLUMNS
    }


@app.post(
    "/predict",
    response_model=PredictionResponse
)
def predict(request: PredictionRequest):

    try:

        input_data = request.model_dump()

        equipment_id = input_data.pop(
            "equipment_id",
            None
        )

        train = input_data.pop(
            "train",
            None
        )

        result = model_service.predict(
            input_data
        )

        return {
            "equipment_id": equipment_id,
            "train": train,
            "failure_probability_24h":
                result["failure_probability_24h"],
            "failure_risk":
                result["failure_risk"],
            "rul_days":
                result["rul_days"],
            "classification_model":
                result["classification_model"],
            "regression_model":
                result["regression_model"]
        }

    except ValueError as exc:

        raise HTTPException(
            status_code=422,
            detail=str(exc)
        )

    except Exception as exc:

        raise HTTPException(
            status_code=500,
            detail="Prediction service error: " + str(exc)
        )


@app.post("/predict/batch")
def predict_batch(
    requests: List[PredictionRequest]
):

    if not requests:

        raise HTTPException(
            status_code=400,
            detail="No prediction records supplied."
        )

    results = []

    for request in requests:

        try:

            input_data = request.model_dump()

            equipment_id = input_data.pop(
                "equipment_id",
                None
            )

            train = input_data.pop(
                "train",
                None
            )

            result = model_service.predict(
                input_data
            )

            results.append({
                "equipment_id": equipment_id,
                "train": train,
                "failure_probability_24h":
                    result["failure_probability_24h"],
                "failure_risk":
                    result["failure_risk"],
                "rul_days":
                    result["rul_days"],
                "classification_model":
                    result["classification_model"],
                "regression_model":
                    result["regression_model"]
            })

        except Exception as exc:

            raise HTTPException(
                status_code=422,
                detail="Batch record failed: " + str(exc)
            )

    return {
        "count": len(results),
        "predictions": results
    }
'''

app_path = os.path.join(
    BACKEND_DIR,
    "app.py"
)

with open(
    app_path,
    "w",
    encoding="utf-8"
) as f:
    f.write(fastapi_app)


# -------------------------------------------------------------------------
# COPY/CREATE the validated model service
# -------------------------------------------------------------------------

model_service_code = '''
import os

import joblib
import numpy as np
import pandas as pd
import tensorflow as tf


BASE_DIR = os.path.dirname(
    os.path.dirname(
        os.path.abspath(__file__)
    )
)

PROJECT_ROOT = os.path.dirname(
    os.path.dirname(
        BASE_DIR
    )
)

MODELS_DIR = os.path.join(
    PROJECT_ROOT,
    "models"
)

CLASSIFICATION_DIR = os.path.join(
    MODELS_DIR,
    "classification"
)

REGRESSION_DIR = os.path.join(
    MODELS_DIR,
    "regression"
)

PREPROCESSING_DIR = os.path.join(
    MODELS_DIR,
    "preprocessing"
)


MLP_MODEL_PATH = os.path.join(
    CLASSIFICATION_DIR,
    "dl_classifier_mlp.keras"
)

MLP_IMPUTER_PATH = os.path.join(
    PREPROCESSING_DIR,
    "dl_feature_imputer.pkl"
)

MLP_SCALER_PATH = os.path.join(
    PREPROCESSING_DIR,
    "dl_feature_scaler.pkl"
)

GB_REGRESSOR_PATH = os.path.join(
    REGRESSION_DIR,
    "gradient_boosting_regressor.pkl"
)


FEATURE_COLUMNS = [
    "commission_year",
    "asset_age_years",
    "hours_since_maint",
    "cumulative_op_hours",
    "bearing_temperature",
    "rpm",
    "vibration",
    "overall_vibration",
    "motor_current",
    "oil_pressure",
    "oil_particles_ppm",
    "bearing_index",
    "discharge_pressure",
    "feed_gas_pressure",
    "lng_output_tph",
    "ambient_temperature",
    "load_factor",
    "wear_level",
    "lubrication_health_index",
    "production_efficiency",
    "quality_factor"
]


class ModelService:

    def __init__(self):

        self.classifier = None
        self.feature_imputer = None
        self.feature_scaler = None
        self.regressor = None
        self.loaded = False

        self.load_models()


    def load_models(self):

        self.classifier = tf.keras.models.load_model(
            MLP_MODEL_PATH,
            compile=False
        )

        self.feature_imputer = joblib.load(
            MLP_IMPUTER_PATH
        )

        self.feature_scaler = joblib.load(
            MLP_SCALER_PATH
        )

        self.regressor = joblib.load(
            GB_REGRESSOR_PATH
        )

        self.loaded = True


    def validate_features(self, data):

        missing = [
            feature
            for feature in FEATURE_COLUMNS
            if feature not in data
        ]

        if missing:

            raise ValueError(
                "Missing required features: "
                + ", ".join(missing)
            )


    def prepare_classifier_input(self, data):

        self.validate_features(data)

        X = pd.DataFrame(
            [[
                data[feature]
                for feature in FEATURE_COLUMNS
            ]],
            columns=FEATURE_COLUMNS
        )

        X_imputed = self.feature_imputer.transform(X)

        X_scaled = self.feature_scaler.transform(
            X_imputed
        )

        return X_scaled.astype(
            np.float32
        )


    def predict_failure(self, data):

        X_scaled = self.prepare_classifier_input(
            data
        )

        probability = float(
            self.classifier.predict(
                X_scaled,
                verbose=0
            ).reshape(-1)[0]
        )

        probability = max(
            0.0,
            min(
                1.0,
                probability
            )
        )

        if probability >= 0.70:
            risk_level = "CRITICAL"

        elif probability >= 0.30:
            risk_level = "HIGH"

        elif probability >= 0.10:
            risk_level = "MODERATE"

        else:
            risk_level = "LOW"

        return {
            "failure_probability_24h": probability,
            "failure_risk": risk_level
        }


    def predict_rul(self, data):

        self.validate_features(data)

        X = pd.DataFrame(
            [[
                data[feature]
                for feature in FEATURE_COLUMNS
            ]],
            columns=FEATURE_COLUMNS
        )

        prediction = float(
            np.asarray(
                self.regressor.predict(X)
            ).reshape(-1)[0]
        )

        prediction = max(
            0.0,
            prediction
        )

        return {
            "rul_days": prediction
        }


    def predict(self, data):

        failure_result = self.predict_failure(
            data
        )

        rul_result = self.predict_rul(
            data
        )

        return {
            **failure_result,
            **rul_result,
            "classification_model":
                "Deep Learning MLP",
            "regression_model":
                "Gradient Boosting Regressor"
        }


model_service = ModelService()
'''

model_service_path = os.path.join(
    SERVICES_DIR,
    "model_service.py"
)

with open(
    model_service_path,
    "w",
    encoding="utf-8"
) as f:
    f.write(model_service_code)


# -------------------------------------------------------------------------
# schemas.py
# -------------------------------------------------------------------------

schemas_code = '''
from typing import Optional

from pydantic import BaseModel, Field


class PredictionRequest(BaseModel):

    equipment_id: Optional[str] = None
    train: Optional[str] = None

    commission_year: float
    asset_age_years: float
    hours_since_maint: float
    cumulative_op_hours: float
    bearing_temperature: float
    rpm: float
    vibration: float
    overall_vibration: float
    motor_current: float
    oil_pressure: float
    oil_particles_ppm: float
    bearing_index: float
    discharge_pressure: float
    feed_gas_pressure: float
    lng_output_tph: float
    ambient_temperature: float
    load_factor: float
    wear_level: float
    lubrication_health_index: float
    production_efficiency: float
    quality_factor: float


class PredictionResponse(BaseModel):

    equipment_id: Optional[str] = None
    train: Optional[str] = None

    failure_probability_24h: float = Field(
        ge=0.0,
        le=1.0
    )

    failure_risk: str
    rul_days: float
    classification_model: str
    regression_model: str


class HealthResponse(BaseModel):

    status: str
    classification_model: str
    regression_model: str
    feature_count: int
'''

schemas_path = os.path.join(
    SCHEMAS_DIR,
    "schemas.py"
)

with open(
    schemas_path,
    "w",
    encoding="utf-8"
) as f:
    f.write(schemas_code)


# -------------------------------------------------------------------------
# __init__.py files
# -------------------------------------------------------------------------

for path in [
    BACKEND_DIR,
    SERVICES_DIR,
    SCHEMAS_DIR
]:

    init_path = os.path.join(
        path,
        "__init__.py"
    )

    if not os.path.exists(init_path):

        with open(
            init_path,
            "w",
            encoding="utf-8"
        ) as f:
            f.write("")


print("=" * 80)
print("FINAL FASTAPI BACKEND CREATED")
print("=" * 80)

print("\nBackend:")
print(app_path)

print("\nModel service:")
print(model_service_path)

print("\nSchemas:")
print(schemas_path)

print("\nExisting Streamlit prototype preserved at:")
print(
    os.path.join(
        PROJECT_ROOT,
        "data",
        "04_corrected_pipeline",
        "idss_backend",
        "app.py"
    )
)

print("\nNext step:")
print(
    r"Run FastAPI from: "
    r"D:\Data_Workspace\Capstone_Project\NLGN_PLC_Predictive_Analystic\idss\backend"
)

print("=" * 80)
```

```python
# =============================================================================
# STEP 10 — CREATE FINAL STREAMLIT IDSS
# =============================================================================

import os

PROJECT_ROOT = (
    r"D:\Data_Workspace\Capstone_Project"
    r"\NLGN_PLC_Predictive_Analystic"
)

DASHBOARD_DIR = os.path.join(
    PROJECT_ROOT,
    "idss",
    "dashboard"
)

os.makedirs(
    DASHBOARD_DIR,
    exist_ok=True
)

DASHBOARD_PATH = os.path.join(
    DASHBOARD_DIR,
    "app.py"
)

dashboard_code = '''
import os
import requests
import pandas as pd
import numpy as np
import streamlit as st
import plotly.graph_objects as go


# =============================================================================
# CONFIGURATION
# =============================================================================

st.set_page_config(
    page_title="NLNG Predictive Maintenance IDSS",
    page_icon="⚙️",
    layout="wide",
    initial_sidebar_state="expanded"
)


PROJECT_ROOT = (
    r"D:\\Data_Workspace\\Capstone_Project"
    r"\\NLGN_PLC_Predictive_Analystic"
)

CORRECTED_DIR = os.path.join(
    PROJECT_ROOT,
    "data",
    "04_corrected_pipeline"
)

STAGE2_PATH = os.path.join(
    CORRECTED_DIR,
    "stage_2_cleaning",
    "NLNG_cleaned_leakage_controlled.csv"
)

SHAP_PATH = os.path.join(
    CORRECTED_DIR,
    "stage_5_explainability",
    "shap_feature_importance.csv"
)

API_URL = "http://127.0.0.1:8000"


# =============================================================================
# AUTHORITATIVE FEATURE CONTRACT
# =============================================================================

FEATURE_COLUMNS = [
    "commission_year",
    "asset_age_years",
    "hours_since_maint",
    "cumulative_op_hours",
    "bearing_temperature",
    "rpm",
    "vibration",
    "overall_vibration",
    "motor_current",
    "oil_pressure",
    "oil_particles_ppm",
    "bearing_index",
    "discharge_pressure",
    "feed_gas_pressure",
    "lng_output_tph",
    "ambient_temperature",
    "load_factor",
    "wear_level",
    "lubrication_health_index",
    "production_efficiency",
    "quality_factor"
]


DASHBOARD_COLUMNS = [
    "timestamp",
    "train",
    "equipment_id",
    "equipment_name",
    "equipment_type",
    "operating_state",
    "criticality"
] + FEATURE_COLUMNS


# =============================================================================
# DATA LOADING
# =============================================================================

@st.cache_data
def load_stage2_data():

    df = pd.read_csv(
        STAGE2_PATH,
        usecols=DASHBOARD_COLUMNS
    )

    df["timestamp"] = pd.to_datetime(
        df["timestamp"],
        errors="coerce"
    )

    return df.sort_values(
        ["equipment_id", "timestamp"]
    )


@st.cache_data
def load_shap_data():

    if not os.path.exists(SHAP_PATH):
        return pd.DataFrame()

    return pd.read_csv(
        SHAP_PATH
    )


# =============================================================================
# API FUNCTIONS
# =============================================================================

def check_api():

    try:

        response = requests.get(
            f"{API_URL}/health",
            timeout=10
        )

        if response.status_code == 200:
            return True, response.json()

        return False, response.text

    except Exception as exc:

        return False, str(exc)


def get_prediction(row):

    payload = {
        "equipment_id": str(row["equipment_id"]),
        "train": str(row["train"])
    }

    for feature in FEATURE_COLUMNS:

        value = row[feature]

        if pd.isna(value):
            payload[feature] = None
        else:
            payload[feature] = float(value)

    response = requests.post(
        f"{API_URL}/predict",
        json=payload,
        timeout=120
    )

    if response.status_code != 200:

        raise RuntimeError(
            f"Prediction API returned "
            f"{response.status_code}: "
            f"{response.text}"
        )

    return response.json()


# =============================================================================
# LOAD DATA
# =============================================================================

try:

    df = load_stage2_data()

except Exception as exc:

    st.error(
        f"Unable to load Stage 2 dataset:\\n\\n{exc}"
    )

    st.stop()


shap_df = load_shap_data()


# =============================================================================
# HEADER
# =============================================================================

st.title(
    "🛡️ NLNG Predictive Maintenance IDSS"
)

st.caption(
    "Pipeline 2.0 model-serving and maintenance "
    "decision-support interface"
)


# =============================================================================
# SIDEBAR
# =============================================================================

st.sidebar.title(
    "🎛️ IDSS Control Center"
)

api_ok, api_info = check_api()

if api_ok:
    st.sidebar.success("Model API: ONLINE")
else:
    st.sidebar.error("Model API: OFFLINE")

st.sidebar.divider()


trains = sorted(
    df["train"].dropna().unique()
)

selected_train = st.sidebar.selectbox(
    "LNG Train",
    trains
)


train_df = df[
    df["train"] == selected_train
]


assets = sorted(
    train_df["equipment_id"]
    .dropna()
    .unique()
)


selected_asset = st.sidebar.selectbox(
    "Equipment",
    assets
)


asset_history = (
    train_df[
        train_df["equipment_id"] == selected_asset
    ]
    .sort_values("timestamp")
    .copy()
)


if asset_history.empty:

    st.error(
        "No observations available for this equipment."
    )

    st.stop()


asset_current = asset_history.iloc[-1]


# =============================================================================
# PREDICTION
# =============================================================================

prediction = None

if api_ok:

    try:

        prediction = get_prediction(
            asset_current
        )

    except Exception as exc:

        st.error(
            f"Prediction failed:\\n\\n{exc}"
        )


# =============================================================================
# TOP KPIs
# =============================================================================

if prediction is not None:

    failure_probability = (
        prediction["failure_probability_24h"]
    )

    failure_percent = (
        failure_probability * 100
    )

    failure_risk = (
        prediction["failure_risk"]
    )

    rul_days = (
        prediction["rul_days"]
    )

else:

    failure_percent = np.nan
    failure_risk = "UNAVAILABLE"
    rul_days = np.nan


col1, col2, col3, col4, col5 = st.columns(5)


with col1:
    st.metric(
        "Equipment",
        selected_asset
    )

with col2:
    st.metric(
        "Train",
        selected_train
    )

with col3:
    st.metric(
        "24h Failure Risk",
        (
            f"{failure_percent:.2f}%"
            if prediction is not None
            else "N/A"
        )
    )

with col4:
    st.metric(
        "Estimated RUL",
        (
            f"{rul_days:.2f} days"
            if prediction is not None
            else "N/A"
        )
    )

with col5:
    st.metric(
        "Operating State",
        str(
            asset_current[
                "operating_state"
            ]
        )
    )


st.divider()


# =============================================================================
# ASSET INFORMATION
# =============================================================================

st.subheader(
    "Equipment Information"
)

i1, i2, i3, i4 = st.columns(4)

with i1:
    st.write("**Equipment Name**")
    st.write(
        asset_current["equipment_name"]
    )

with i2:
    st.write("**Equipment Type**")
    st.write(
        asset_current["equipment_type"]
    )

with i3:
    st.write("**Criticality**")
    st.write(
        asset_current["criticality"]
    )

with i4:
    st.write("**Latest Observation**")
    st.write(
        asset_current["timestamp"].strftime(
            "%Y-%m-%d %H:%M"
        )
    )


st.divider()


# =============================================================================
# CONDITION TRENDS + RISK GAUGE
# =============================================================================

left, right = st.columns(
    [1.7, 1]
)


with left:

    st.subheader(
        "📈 Equipment Condition Trends"
    )

    recent = (
        asset_history
        .tail(500)
        .copy()
    )

    fig = go.Figure()

    fig.add_trace(
        go.Scatter(
            x=recent["timestamp"],
            y=recent["overall_vibration"],
            name="Overall Vibration",
            mode="lines"
        )
    )

    fig.add_trace(
        go.Scatter(
            x=recent["timestamp"],
            y=recent["oil_particles_ppm"],
            name="Oil Particles",
            mode="lines"
        )
    )

    fig.update_layout(
        title="Recent Equipment Condition",
        xaxis_title="Timestamp",
        yaxis_title="Value",
        template="plotly_dark",
        height=400
    )

    st.plotly_chart(
        fig,
        width="stretch"
    )


with right:

    st.subheader(
        "🎯 24-Hour Failure Risk"
    )

    gauge_value = (
        failure_percent
        if prediction is not None
        else 0
    )

    fig = go.Figure(
        go.Indicator(
            mode="gauge+number",
            value=gauge_value,
            title={
                "text":
                    "Failure Probability (%)"
            },
            gauge={
                "axis": {
                    "range": [0, 100]
                },
                "steps": [
                    {
                        "range": [0, 10],
                        "color": "#00CC96"
                    },
                    {
                        "range": [10, 30],
                        "color": "#FFAA00"
                    },
                    {
                        "range": [30, 100],
                        "color": "#FF4B4B"
                    }
                ]
            }
        )
    )

    fig.update_layout(
        template="plotly_dark",
        height=400
    )

    st.plotly_chart(
        fig,
        width="stretch"
    )

    if prediction is not None:

        if failure_risk == "LOW":
            st.success(
                f"🟢 Risk Level: {failure_risk}"
            )

        elif failure_risk == "MODERATE":
            st.warning(
                f"🟡 Risk Level: {failure_risk}"
            )

        elif failure_risk == "HIGH":
            st.warning(
                f"🟠 Risk Level: {failure_risk}"
            )

        else:
            st.error(
                f"🔴 Risk Level: {failure_risk}"
            )


st.divider()


# =============================================================================
# RUL
# =============================================================================

st.subheader(
    "⏳ Remaining Useful Life"
)

if prediction is not None:

    st.metric(
        "Estimated RUL",
        f"{rul_days:.2f} days"
    )

    st.info(
        "RUL is presented as a model-generated estimate "
        "for maintenance planning support and should not "
        "be interpreted as an exact failure date."
    )

else:

    st.warning(
        "RUL prediction unavailable."
    )


st.divider()


# =============================================================================
# SERVED MODEL INFORMATION
# =============================================================================

st.subheader(
    "🤖 Models Being Served"
)

m1, m2 = st.columns(2)

with m1:
    st.info(
        "**Classification**\\n\\n"
        "Deep Learning MLP\\n\\n"
        "24-hour failure-risk prediction"
    )

with m2:
    st.info(
        "**Regression**\\n\\n"
        "Gradient Boosting Regressor\\n\\n"
        "Remaining Useful Life estimation"
    )


st.divider()


# =============================================================================
# SHAP GLOBAL DRIVERS
# =============================================================================

st.subheader(
    "🔎 Global Model Drivers"
)

if not shap_df.empty:

    top = (
        shap_df
        .head(10)
        .copy()
    )

    fig = go.Figure(
        go.Bar(
            x=top["Mean_Absolute_SHAP"],
            y=top["Feature"],
            orientation="h"
        )
    )

    fig.update_layout(
        title="Top Global Failure-Prediction Drivers",
        xaxis_title="Mean Absolute SHAP Value",
        yaxis_title="Feature",
        template="plotly_dark",
        height=450,
        yaxis={
            "categoryorder": "total ascending"
        }
    )

    st.plotly_chart(
        fig,
        width="stretch"
    )

    st.caption(
        "These are global model drivers from the Stage 5 "
        "SHAP analysis. They describe overall model behaviour "
        "and are not a case-specific causal explanation."
    )


st.divider()


# =============================================================================
# DECISION SUPPORT
# =============================================================================

st.subheader(
    "💡 Predictive Maintenance Decision Support"
)

if prediction is None:

    st.info(
        "Prediction unavailable. Start the FastAPI "
        "model-serving service."
    )

else:

    if failure_risk == "CRITICAL":

        st.error(
            f"""
            **CRITICAL RISK**

            {selected_asset} has a high predicted probability
            of failure within the defined 24-hour horizon.

            **Suggested decision-support action:**
            Prioritise engineering assessment and condition review.
            """
        )

    elif failure_risk == "HIGH":

        st.warning(
            f"""
            **HIGH RISK**

            {selected_asset} requires increased monitoring
            and maintenance review.

            **Suggested decision-support action:**
            Prioritise condition assessment and maintenance planning.
            """
        )

    elif failure_risk == "MODERATE":

        st.warning(
            f"""
            **MODERATE RISK**

            {selected_asset} shows elevated predicted failure risk.

            **Suggested decision-support action:**
            Review current condition indicators and continue
            focused monitoring.
            """
        )

    else:

        st.success(
            f"""
            **LOW RISK**

            {selected_asset} has a low predicted probability
            of failure within the defined 24-hour horizon.

            **Suggested decision-support action:**
            Continue normal condition monitoring.
            """
        )

    st.caption(
        "The IDSS provides decision-support information. "
        "It does not automatically initiate maintenance, "
        "shutdown, or plant-control actions."
    )


st.divider()

st.caption(
    "NLNG Predictive Maintenance IDSS | "
    "Streamlit → FastAPI → Pipeline 2.0 Models"
)
'''

with open(
    DASHBOARD_PATH,
    "w",
    encoding="utf-8"
) as f:
    f.write(dashboard_code)

print("=" * 80)
print("STEP 10 COMPLETE")
print("=" * 80)
print("Dashboard created:")
print(DASHBOARD_PATH)
```
